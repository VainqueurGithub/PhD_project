PLOT DE RELATION BETWEEN VERTEBRATE RICHNESS AND TERRAIN PORPERTIES

```python
import matplotlib.pyplot as plt
from scipy.stats import pearsonr
import numpy as np

terrain_variables = {
    "slope_mean": "Slope mean",
    "slope_std": "Slope variability",
    "TPI_mean": "TPI mean",
    "TPI_std": "TPI variability",
    "TRI_mean": "TRI mean",
    "TRI_std": "TRI variability",
    "roughness_mean": "Roughness mean",
    "roughness_std": "Roughness variability",
    "hillshade_mean": "Hillshade mean",
    "hillshade_std": "Hillshade variability"
}

fig, axes = plt.subplots(2,5, figsize=(20,10))
axes = axes.flatten()

for ax, (variable, label) in zip(axes, terrain_variables.items()):
    # Removing missing value
    data = topography[["richness", variable]].dropna()
    x = data[variable]
    y = data["richness"]
    # Scatter plot
    ax.scatter(x,y, alpha=0.5, s=25)
    # Regression line
    if len(data)>1:
        slope, intercept = np.polyfit(x,y,1)
        x_line = np.linspace(x.min(), x.max(), 100)
        y_line = slope * x_line + intercept

        ax.plot(x_line, y_line, linewidth=2)

        # Pearson correlation
        r, p = pearsonr(x, y)

        ax.text(0.05,0.95,f"r = {r:.2f}\np = {p:.3g}",transform=ax.transAxes,verticalalignment="top",fontweight="bold",fontsize=11)
    # Bold axis labels
    ax.set_xlabel(
        label,
        fontweight="bold",
        fontsize=12
    )

    ax.set_ylabel(
        "Species richness",
        fontweight="bold",
        fontsize=12
    )

    # Bold tick labels
    ax.tick_params(axis="both", labelsize=10)

    for tick in ax.get_xticklabels():
        tick.set_fontweight("bold")

    for tick in ax.get_yticklabels():
        tick.set_fontweight("bold")

    ax.grid(alpha=0.2)

plt.tight_layout()

# Save as high-resolution PNG
plt.savefig(
    "species_richness_terrain_relationships.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

SPECIES RICHNESS VS TERRAIN MAPS 

```python
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
from matplotlib.gridspec import GridSpec
from sklearn.preprocessing import StandardScaler

terrain_variables = {
    "slope_mean": "Slope mean",
    "slope_std": "Slope variability",
    "TPI_mean": "TPI mean",
    "TPI_std": "TPI variability",
    "TRI_mean": "TRI mean",
    "TRI_std": "TRI variability",
    "roughness_mean": "Roughness mean",
    "roughness_std": "Roughness variability",
    "hillshade_mean": "Hillshade mean",
    "hillshade_std": "Hillshade variability"
}

variables = ["richness"] + list(terrain_variables.keys())

# -----------------------------
# Standardize
# -----------------------------

df_std = topography.copy()

df_std[variables] = StandardScaler().fit_transform(topography[variables])

# -----------------------------
# Common colour scale
# -----------------------------

norm = mcolors.TwoSlopeNorm(vmin=-2,vcenter=0,vmax=2)

# -----------------------------
# Figure layout
# -----------------------------

fig = plt.figure(figsize=(19, 13))

gs = GridSpec(
    3,
    5,
    figure=fig,
    width_ratios=[1, 1, 1, 1, 0.08],
    wspace=0.15,
    hspace=0.15
)

# Map axes
axes = []

for row in range(3):
    for col in range(4):
        axes.append(
            fig.add_subplot(gs[row, col])
        )

# Colourbar axis
cbar_ax = fig.add_subplot(gs[:, 4])


# -----------------------------
# Species richness
# -----------------------------

df_std.plot(
    column="richness",
    ax=axes[0],
    cmap="RdBu_r",
    norm=norm,
    linewidth=0
)

axes[0].set_title(
    "Species richness",
    fontsize=13,
    fontweight="bold"
)

axes[0].set_axis_off()


# -----------------------------
# Terrain variables
# -----------------------------

for ax, (variable, label) in zip(
    axes[1:],
    terrain_variables.items()
):

    df_std.plot(
        column=variable,
        ax=ax,
        cmap="RdBu_r",
        norm=norm,
        linewidth=0
    )

    ax.set_title(
        label,
        fontsize=12,
        fontweight="bold"
    )

    ax.set_axis_off()


# Hide unused 12th map
axes[-1].set_visible(False)


# -----------------------------
# Colourbar
# -----------------------------

sm = plt.cm.ScalarMappable(
    cmap="RdBu_r",
    norm=norm
)

sm.set_array([])

cbar = fig.colorbar(
    sm,
    cax=cbar_ax
)

cbar.set_label(
    "Standardized value (z-score)",
    fontsize=12,
    fontweight="bold"
)

cbar.ax.tick_params(labelsize=10)

for tick in cbar.ax.get_yticklabels():
    tick.set_fontweight("bold")


# -----------------------------
# Save
# -----------------------------

plt.savefig(
    "standardized_species_richness_terrain_maps.png",
    dpi=600,
    bbox_inches="tight"
)

plt.show()
```
