```python
veg_summary = pd.read_csv("data/frank_white_vegetation_by_grid.csv")
veg_summary
```

```python
import numpy as np
vegetation_heterogeneity = (
    veg_summary
    .groupby("grid")
    .agg(
        vegetation_richness=("MOSAIC", "nunique"),
        vegetation_shannon=("proportion",
                            lambda p: -(p * np.log(p)).sum())
    )
    .reset_index()
)

vegetation_heterogeneity["vegetation_shannon"] = (
    vegetation_heterogeneity["vegetation_shannon"].replace(-0.0, 0.0)
)
```

```python
vegetation_heterogeneity["vegetation_evenness"] = np.where(
    vegetation_heterogeneity["vegetation_richness"] > 1,
    vegetation_heterogeneity["vegetation_shannon"]
    / np.log(vegetation_heterogeneity["vegetation_richness"]),
    1
)
```

```python
df_moisaic_veg_rich = mosaic_vegetation.merge(
    vegetation_heterogeneity,
    on="grid",
    how="left"
)
```

RICHNESS PATTERNS ANALYSIS

OVERALL RICHNESS VS MOSAIC VEGETATION CLASSIFICATION PROPORTION

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import pearsonr

veg_wide = mosaic_vegetation[['grid','APSWG','ESGKS','GCRF-D','ITA','LRF-W','MANG','MOS-1A2','MOS-8A1A','MOS-EAB-AWG','MOS-LRF-IW-SG','MOS-LRF-SG',
                              'MOS-WZW-SG','OUTSIDE','SDGS','SW-I','SWAMP','TRF','UMV','UW','WACM','WATER','WZMW']]

vegetation_columns = [
    col for col in veg_wide.columns
    if col != "grid"
]

# ---------------------------------------------------------
# Create figure
# ---------------------------------------------------------

fig, axes = plt.subplots(
    nrows=5,
    ncols=5,
    figsize=(20, 20)
)

axes = axes.flatten()

# ---------------------------------------------------------
# Plot each vegetation type
# ---------------------------------------------------------

for ax, veg in zip(axes, vegetation_columns):

    temp = mosaic_vegetation[
        [veg, "richness"]
    ].dropna()

    # Pearson correlation
    if len(temp) >= 3 and temp[veg].nunique() > 1:
        r, p = pearsonr(
            temp[veg],
            temp["richness"]
        )
    else:
        r, p = np.nan, np.nan

    # Regression plot
    sns.regplot(
        data=temp,
        x=veg,
        y="richness",
        scatter_kws={
            "alpha": 0.15,
            "s": 10
        },
        line_kws={
            "linewidth": 2
        },
        ax=ax
    )

    # Axis labels
    ax.set_xlabel(
        f"{veg} (%)",
        fontweight="bold"
    )

    ax.set_ylabel(
        "Species richness",
        fontweight="bold"
    )

    # Short vegetation name
    ax.set_title(
        veg,
        fontweight="bold",
        fontsize=12
    )

    # r and p
    ax.text(
        0.05,
        0.95,
        f"$r$ = {r:.2f}\n$p$ = {p:.2e}",
        transform=ax.transAxes,
        ha="left",
        va="top",
        fontsize=10,
        fontweight="bold"
    )

    ax.grid(
        alpha=0.2
    )

# ---------------------------------------------------------
# Remove unused axes
# ---------------------------------------------------------

for ax in axes[len(vegetation_columns):]:
    ax.remove()

# ---------------------------------------------------------
# Overall title
# ---------------------------------------------------------

fig.suptitle(
    "Species richness versus vegetation proportion",
    fontsize=18,
    fontweight="bold",
    y=0.995
)

plt.tight_layout(
    rect=[0, 0.20, 1, 0.98]
)

plt.savefig(
    "species_richness_vs_vegetation_proportion.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

OVERALL RICHNESS VS MOSAIC VEGETATION RICHNESS, SHANNON DIVERSITY, AND EVENNESS

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import pearsonr, linregress

vegetation_variables = {
    "vegetation_richness": "Vegetation richness",
    "vegetation_shannon": "Vegetation Shannon diversity",
    "vegetation_evenness": "Vegetation evenness"
}

fig, axes = plt.subplots(
    1, 3,
    figsize=(18, 5)
)

for ax, (x_var, x_label) in zip(axes, vegetation_variables.items()):

    # Keep only complete observations
    data = mosaic_vegetation[["richness", x_var]].dropna()

    x = data[x_var]
    y = data["richness"]

    # Pearson correlation
    r, p = pearsonr(x, y)

    # Linear regression
    slope, intercept, _, _, _ = linregress(x, y)

    # Scatter points
    ax.scatter(
        x,
        y,
        s=25,
        alpha=0.5
    )

    # Regression line
    x_line = np.linspace(x.min(), x.max(), 100)
    y_line = intercept + slope * x_line

    ax.plot(
        x_line,
        y_line,
        linewidth=2
    )

    # r and p
    ax.text(
        0.05,
        0.95,
        f"$r$ = {r:.2f}\n$p$ = {p:.3g}",
        transform=ax.transAxes,
        fontsize=12,
        fontweight="bold",
        verticalalignment="top"
    )

    # Axis labels
    ax.set_xlabel(
        x_label,
        fontsize=12,
        fontweight="bold"
    )

    ax.set_ylabel(
        "Species richness",
        fontsize=12,
        fontweight="bold"
    )

    # Bold tick labels
    for tick in ax.get_xticklabels() + ax.get_yticklabels():
        tick.set_fontweight("bold")

    # Border
    for spine in ax.spines.values():
        spine.set_linewidth(1.2)

plt.tight_layout()

plt.savefig(
    "species_richness__mosaic_vegetation_heterogeneity.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import pearsonr, linregress

vegetation_variables = {
    "vegetation_richness": "Vegetation richness",
    "vegetation_shannon": "Vegetation Shannon diversity",
    "vegetation_evenness": "Vegetation evenness"
}

taxon_dataset=[birds, mammals, amphibians, reptiles]
label = ["Birds", "Mammals", "Amphibians", "Reptiles"]

for index, taxon in enumerate(taxon_dataset):
    fig, axes = plt.subplots(1, 3,figsize=(18, 5))
    
    for ax, (x_var, x_label) in zip(axes, vegetation_variables.items()):
        # Keep only complete observations
        data = taxon[["richness", x_var]].dropna()
        x = data[x_var]
        y = data["richness"]
        # Pearson correlation
        r, p = pearsonr(x, y)
        # Linear regression
        slope, intercept, _, _, _ = linregress(x, y)
        # Scatter points
        ax.scatter(x,y,s=25,alpha=0.5)
        # Regression line
        x_line = np.linspace(x.min(), x.max(), 100)
        y_line = intercept + slope * x_line
        ax.plot(x_line,y_line,linewidth=2)
        # r and p
        ax.text(0.05,0.95,f"$r$ = {r:.2f}\n$p$ = {p:.3g}",transform=ax.transAxes,fontsize=12,fontweight="bold",verticalalignment="top")
        # Axis labels
        ax.set_xlabel(x_label,fontsize=12,fontweight="bold")
        ax.set_ylabel(label[index]+" richness",fontsize=12,fontweight="bold")
        # Bold tick labels
        for tick in ax.get_xticklabels() + ax.get_yticklabels():
            tick.set_fontweight("bold")

        # Border
        for spine in ax.spines.values():
            spine.set_linewidth(1.2)
    plt.tight_layout()
    plt.savefig(label[index]+"_richness__mosaic_vegetation_heterogeneity.png",dpi=300,bbox_inches="tight")
    plt.show()
```

OVERALL WEIGHTED ENDEMISM AND COMMUNITY WEIGHTED ENDEMISM VS MOSAIC VEGETATION CLASSIFICATION PROPORTION

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import pearsonr

veg_wide = mosaic_vegetation[['grid','APSWG','ESGKS','GCRF-D','ITA','LRF-W','MANG','MOS-1A2','MOS-8A1A','MOS-EAB-AWG','MOS-LRF-IW-SG','MOS-LRF-SG',
                              'MOS-WZW-SG','OUTSIDE','SDGS','SW-I','SWAMP','TRF','UMV','UW','WACM','WATER','WZMW']]

vegetation_columns = [
    col for col in veg_wide.columns
    if col != "grid"
]

# ---------------------------------------------------------
# Create figure
# ---------------------------------------------------------

fig, axes = plt.subplots(
    nrows=5,
    ncols=5,
    figsize=(20, 20)
)

axes = axes.flatten()

# ---------------------------------------------------------
# Plot each vegetation type
# ---------------------------------------------------------

for ax, veg in zip(axes, vegetation_columns):

    temp = mosaic_vegetation[[veg, "WE", "CWE"]].dropna()

    # Pearson correlation WE
    if len(temp) >= 3 and temp[veg].nunique() > 1:
        r_we, p_we = pearsonr(temp[veg],temp["WE"])
    else:
        r_we, p_we = np.nan, np.nan

    # Pearson correlation CWE
    if len(temp) >= 3 and temp[veg].nunique() > 1:
        r_cwe, p_cwe = pearsonr(temp[veg],temp["CWE"])
    else:
        r_cwe, p_cwe = np.nan, np.nan

    # WE Regression plot
    sns.regplot(data=temp,x=veg,y="WE",scatter_kws={"alpha": 0.12,"s": 10},line_kws={"linewidth": 2,"linestyle": "-"},ax=ax)
    # CWE Regression plot
    sns.regplot(data=temp,x=veg,y="CWE",scatter_kws={"alpha": 0.12,"s": 10},line_kws={"linewidth": 2,"linestyle": "--"},ax=ax)
    # Axis labels
    ax.set_xlabel(f"{veg} (%)",fontweight="bold")

    ax.set_ylabel("Endemism",fontweight="bold")

    # Short vegetation name
    ax.set_title(veg,fontweight="bold",fontsize=12)

    # r and p
    ax.text(0.05,0.95,f"WE:  $r$ = {r_we:.2f}, $p$ = {p_we:.2e}\n"f"CWE: $r$ = {r_cwe:.2f}, $p$ = {p_cwe:.2e}",transform=ax.transAxes,ha="left",
        va="top",fontsize=9,fontweight="bold")

    ax.grid(
        alpha=0.2
    )

# ---------------------------------------------------------
# Remove unused axes
# ---------------------------------------------------------

for ax in axes[len(vegetation_columns):]:
    ax.remove()

# ---------------------------------------------------------
# Overall title
# ---------------------------------------------------------

fig.suptitle(
    "Weighted endemism (WE) and corrected weighted endemism (CWE)",
    fontsize=18,
    fontweight="bold",
    y=0.995
)

# ---------------------------------------------------------
# Legend
# ---------------------------------------------------------

from matplotlib.lines import Line2D

legend_elements = [
    Line2D(
        [0], [0],
        linestyle="-",
        linewidth=2,
        label="WE"
    ),
    Line2D(
        [0], [0],
        linestyle="--",
        linewidth=2,
        label="CWE"
    )
]

fig.legend(
    handles=legend_elements,
    loc="lower center",
    ncol=2,
    fontsize=12,
    frameon=False
)

# ---------------------------------------------------------
# Layout
# ---------------------------------------------------------

plt.tight_layout(
    rect=[0, 0.03, 1, 0.97]
)

# ---------------------------------------------------------
# Save
# ---------------------------------------------------------

plt.savefig(
    "WE_CWE_vs_mosaic_vegetation_proportion.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
