```python
from statsmodels.discrete.discrete_model import NegativeBinomial

for vegetation_var in vegetation_vars:

    formula = (
        "richness ~ "
        "C(ORD_STRA) + "
        "distance_km + "
        "LENGTH_KM + "
        "DIS_AV_CMS + "
        f"{vegetation_var} + "
        "slope_std + "
        "TPI_mean + "
        "TRI_mean +"
        "TRI_std +"
        "roughness_std + "
        "hillshade_mean + "
        "hillshade_std + "
        "refugia_dist_km + "
        "BIO1_temporal_mean + "
        "BIO1_temporal_SD + "
        "BIO12_temporal_mean + "
        "BIO12_temporal_SD + "
        "forest_loss_mean + "
        "forest_canopy_mean + "
        "forest_canopy_stdev + "
        "tree_cover_mean + "
        "tree_cover_stdev + "
        "clay_mean + "
        "clay_stdev + "
        "cec_mean + "
        "cec_stdev + "
        "soc_mean + "
        "soc_stdev + "
        "ph_mean + "
        "ph_stdev"
    )

   
    model = NegativeBinomial.from_formula(formula,data=df_model).fit(method="bfgs",maxiter=2000,disp=True)

    richness_models[vegetation_var] = model

    print("\n")
    print("=" * 80)
    print(vegetation_var)
    print("=" * 80)
    print(model.summary())
```

```python
import statsmodels.api as sm
import statsmodels.formula.api as smf

endemism_models = {}

df_model["ORD_STRA"] = grid["ORD_STRA"].astype("category") # Make sure this ORDER_STRA is treaten as categorical

for vegetation_var in vegetation_vars:

    formula = (
        "WE ~ "
        "C(ORD_STRA) + "
        "distance_km + "
        "LENGTH_KM + "
        "DIS_AV_CMS + "
        f"{vegetation_var} + "
        "slope_std + "
        "TPI_mean + "
        "TRI_mean +"
        "TRI_std +"
        "roughness_std + "
        "hillshade_mean + "
        "hillshade_std + "
        "refugia_dist_km + "
        "BIO1_temporal_mean + "
        "BIO1_temporal_SD + "
        "BIO12_temporal_mean + "
        "BIO12_temporal_SD + "
        "forest_loss_mean + "
        "forest_canopy_mean + "
        "forest_canopy_stdev + "
        "tree_cover_mean + "
        "tree_cover_stdev + "
        "clay_mean + "
        "clay_stdev + "
        "cec_mean + "
        "cec_stdev + "
        "soc_mean + "
        "soc_stdev + "
        "ph_mean + "
        "ph_stdev "
    )

    model = smf.glm(formula=formula,data=df_model,family=sm.families.Gamma(link=sm.families.links.Log())).fit(method="lbfgs")
    endemism_models[vegetation_var] = model

    print("\n")
    print("=" * 80)
    print(vegetation_var)
    print("=" * 80)
    print(model.summary())
```

```python
def extract_coefficients(models, response_name):

    results = []

    for vegetation_var, model in models.items():

        params = model.params
        pvalues = model.pvalues
        conf = model.conf_int()

        for variable in params.index:

            if variable == "Intercept":
                continue

            results.append({
                "response": response_name,
                "vegetation_model": vegetation_var,
                "predictor": variable,
                "coefficient": params[variable],
                "p_value": pvalues[variable],
                "CI_lower": conf.loc[variable, 0],
                "CI_upper": conf.loc[variable, 1]
            })

    return pd.DataFrame(results)


richness_results = extract_coefficients(
    richness_models,
    "Species richness"
)

endemism_results = extract_coefficients(
    endemism_models,
    "Species endemism"
)

all_results = pd.concat(
    [richness_results, endemism_results],
    ignore_index=True
)

all_results
```

COEFFICIENT PLOTS FOR BOTH OVERALL RICHNESS AND ENDEMISM

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


# ---------------------------------------------------------
# 1. Function to extract coefficients
# ---------------------------------------------------------

def extract_coefficients(model, vegetation_name, response_name):

    params = model.params
    conf = model.conf_int()

    results = pd.DataFrame({
        "term": params.index,
        "beta": params.values,
        "lower_beta": conf[0].values,
        "upper_beta": conf[1].values
    })

    # Convert log coefficients to IRR
    results["IRR"] = np.exp(results["beta"])
    results["lower_IRR"] = np.exp(results["lower_beta"])
    results["upper_IRR"] = np.exp(results["upper_beta"])
     
    results["vegetation_model"] = vegetation_name
    results["response"] = response_name

    return results


# ---------------------------------------------------------
# 2. Extract richness models
# ---------------------------------------------------------

richness_results = pd.concat([
    extract_coefficients(
        richness_models["mosaic_vegetation_richness"],
        "Vegetation richness",
        "Species richness"
    ),
    extract_coefficients(
        richness_models["mosaic_vegetation_shannon"],
        "Vegetation Shannon",
        "Species richness"
    ),
    extract_coefficients(
        richness_models["mosaic_vegetation_evenness"],
        "Vegetation evenness",
        "Species richness"
    )
], ignore_index=True)


# ---------------------------------------------------------
# 3. Extract endemism models
# ---------------------------------------------------------

endemism_results = pd.concat([
    extract_coefficients(
        endemism_models["mosaic_vegetation_richness"],
        "Vegetation richness",
        "Species endemism"
    ),
    extract_coefficients(
        endemism_models["mosaic_vegetation_richness"],
        "Vegetation Shannon",
        "Species endemism"
    ),
    extract_coefficients(
        endemism_models["mosaic_vegetation_richness"],
        "Vegetation evenness",
        "Species endemism"
    )
], ignore_index=True)


# ---------------------------------------------------------
# 4. Combine
# ---------------------------------------------------------

results = pd.concat(
    [richness_results, endemism_results],
    ignore_index=True
)
results
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import Normalize, TwoSlopeNorm
from matplotlib.cm import ScalarMappable


fig, axes = plt.subplots(2, 3,figsize=(18, 11))


# =========================================================
# DATA
# =========================================================

responses = [
    {
        "name": "Species richness",
        "observed": "richness",
        "predicted": "richness_pred",
        "residual": "richness_residual"
    },
    {
        "name": "Species endemism",
        "observed": "WE",
        "predicted": "endemism_pred",
        "residual": "endemism_residual"
    }
]


# =========================================================
# LOOP
# =========================================================

for row, response in enumerate(responses):

    observed = response["observed"]
    predicted = response["predicted"]
    residual = response["residual"]


    # =====================================================
    # OBSERVED + PREDICTED SCALE
    # =====================================================

    combined = np.concatenate([
        df_model[observed].dropna().values,
        df_model[predicted].dropna().values
    ])

    vmin = np.percentile(combined, 1)
    vmax = np.percentile(combined, 99)


    # =====================================================
    # RESIDUAL SCALE
    # =====================================================

    residual_values = df_model[residual].dropna().values

    residual_max = np.percentile(
        np.abs(residual_values),
        99
    )


    # =====================================================
    # OBSERVED
    # =====================================================

    df_model.plot(
        column=observed,
        ax=axes[row, 0],
        cmap="viridis",
        vmin=vmin,
        vmax=vmax,
        linewidth=0,
        edgecolor="none"
    )

    axes[row, 0].set_title(
        "Observed",
        fontsize=14,
        fontweight="bold"
    )


    # =====================================================
    # PREDICTED
    # =====================================================

    df_model.plot(
        column=predicted,
        ax=axes[row, 1],
        cmap="viridis",
        vmin=vmin,
        vmax=vmax,
        linewidth=0,
        edgecolor="none"
    )

    axes[row, 1].set_title(
        "Predicted",
        fontsize=14,
        fontweight="bold"
    )


    # =====================================================
    # RESIDUAL
    # =====================================================

    df_model.plot(
        column=residual,
        ax=axes[row, 2],
        cmap="RdBu_r",
        vmin=-residual_max,
        vmax=residual_max,
        linewidth=0,
        edgecolor="none"
    )

    axes[row, 2].set_title(
        "Residual",
        fontsize=14,
        fontweight="bold"
    )


    # =====================================================
    # ROW LABEL
    # =====================================================

    axes[row, 0].set_ylabel(
        response["name"],
        fontsize=14,
        fontweight="bold"
    )


    # =====================================================
    # REMOVE AXES
    # =====================================================

    for col in range(3):
        axes[row, col].set_axis_off()


    # =====================================================
    # OBSERVED/PREDICTED COLORBAR
    # =====================================================

    sm = ScalarMappable(
        norm=Normalize(
            vmin=vmin,
            vmax=vmax
        ),
        cmap="viridis"
    )

    sm.set_array([])

    cbar = fig.colorbar(
        sm,
        ax=axes[row, 0:2],
        fraction=0.025,
        pad=0.02
    )

    cbar.set_label(
        response["name"],
        fontsize=11,
        fontweight="bold"
    )


    # =====================================================
    # RESIDUAL COLORBAR
    # =====================================================

    sm_res = ScalarMappable(
        norm=TwoSlopeNorm(
            vmin=-residual_max,
            vcenter=0,
            vmax=residual_max
        ),
        cmap="RdBu_r"
    )

    sm_res.set_array([])

    cbar_res = fig.colorbar(
        sm_res,
        ax=axes[row, 2],
        fraction=0.025,
        pad=0.02
    )

    cbar_res.set_label(
        "Pearson residual",
        fontsize=11,
        fontweight="bold"
    )


# =========================================================
# TITLE
# =========================================================

fig.suptitle(
    "Observed, predicted and residual biodiversity",
    fontsize=18,
    fontweight="bold",
    y=0.98
)

plt.tight_layout(rect=[0, 0, 0.95, 1])

plt.savefig(
    "observed_predicted_residual_biodiversity1.png",
    dpi=600,
    bbox_inches="tight"
)

plt.show()
```

PREDICTOR IMPORTANCE PLOT

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import mean_squared_error


# =========================================================
# PREDICTORS
# =========================================================

predictors = [
    "ORD_STRA",
    "distance_km",
    "LENGTH_KM",
    "DIS_AV_CMS",
    "mosaic_vegetation_richness",
    "mosaic_vegetation_shannon",
    "mosaic_vegetation_evenness",
    "slope_std",
    "TPI_mean",
    "TRI_mean",
    "roughness_std",
    "hillshade_mean",
    "hillshade_std",
    "refugia_dist_km",
    "BIO1_temporal_mean",
    "BIO1_temporal_SD",
    "BIO12_temporal_mean",
    "BIO12_temporal_SD",
    "forest_loss_mean",
    "forest_canopy_mean",
    "forest_canopy_stdev",
    "tree_cover_mean",
    "tree_cover_stdev",
    "clay_mean",
    "clay_stdev",
    "cec_mean",
    "cec_stdev",
    "soc_mean",
    "soc_stdev",
    "ph_mean",
    "ph_stdev"
]


labels = {
    "ORD_STRA": "River order",
    "distance_km": "Distance to river",
    "LENGTH_KM": "River length",
    "DIS_AV_CMS": "Average discharge",
    "mosaic_vegetation_richness": "Vegetation richness",
    "mosaic_vegetation_shannon": "Vegetation Shannon",
    "mosaic_vegetation_evenness": "Vegetation evenness",
    "slope_std": "Slope variability",
    "TPI_mean": "Mean topographic position index",
    "TRI_mean": "Mean terrain ruggedness index",
    "TRI_std": "Terrain ruggedness variability",
    "roughness_std": "Terrain roughness variability",
    "hillshade_mean": "Mean hillshade",
    "hillshade_std": "Hillshade variability",
    "refugia_dist_km": "Distance to nearest refugium (km)",
    "BIO1_temporal_mean": "Mean temperature (°c)",
    "BIO1_temporal_SD": "Temperature variability (°c)",
    "BIO12_temporal_mean": "Mean precipitation (mm)",
    "BIO12_temporal_SD": "Precipitation variability (mm)",
    "forest_loss_mean": "Mean forest loss from 2001 to 2020",
    "forest_canopy_mean": "Mean forest canopy height (m)",
    "forest_canopy_stdev": "Forest canopy height variation (mm)",
    "tree_cover_mean": "Mean tree cover (%)",
    "tree_cover_stdev": "Tree cover variability (%)",
    "clay_mean": "Mean soil clay at 0 to 5 cm",
    "clay_stdev": "Soil clay variability at 0 to 5 cm",
    "cec_mean": "Mean soil cation exchange capacity at 0 to 5 cm",
    "cec_stdev": "Soil cation exchange capacity variability at 0 to 5 cm",
    "soc_mean": "Mean soil organic carbon at 0 to 5 cm",
    "soc_stdev": "Soil organic carbon variability at 0 to 5 cm",
    "ph_mean": "Mean soil PH at 0 to 5 cm",
    "ph_stdev": "Soil PH variability at 0 to 5 cm"
}


# =========================================================
# FUNCTION
# =========================================================

def permutation_importance_rmse(model, data, response,predictors, n_permutations=10,seed=42):

    y = data[response]

    # Remove rows with missing response
    valid = y.notna()

    data = data.loc[valid].copy()
    y = y.loc[valid]

    baseline_pred = model.predict(data)

    baseline_rmse = np.sqrt(
        mean_squared_error(y, baseline_pred)
    )

    rng = np.random.default_rng(seed)

    results = []

    for predictor in predictors:

        errors = []

        for i in range(n_permutations):

            data_perm = data.copy()

            data_perm[predictor] = rng.permutation(
                data_perm[predictor].values
            )

            pred = model.predict(data_perm)

            rmse = np.sqrt(
                mean_squared_error(y, pred)
            )

            errors.append(
                rmse - baseline_rmse
            )

        results.append({
            "predictor": predictor,
            "importance": np.mean(errors),
            "std": np.std(errors)
        })

    return pd.DataFrame(results)


# =========================================================
# RICHNESS
# =========================================================

richness_importance = permutation_importance_rmse(
    model=richness_models['mosaic_vegetation_richness'],
    data=df_model,
    response="richness",
    predictors=predictors,
    n_permutations=10
)


# =========================================================
# ENDEMISM
# =========================================================

endemism_importance = permutation_importance_rmse(
    model=endemism_models['mosaic_vegetation_richness'],
    data=df_model,
    response="WE",
    predictors=predictors,
    n_permutations=10
)
```

```python
# =========================================================
# PREPARE DATA
# =========================================================

richness_importance["label"] = (
    richness_importance["predictor"].map(labels)
)

endemism_importance["label"] = (
    endemism_importance["predictor"].map(labels)
)


# Use the same predictor order in both panels
order = richness_importance.sort_values(
    "importance"
)["predictor"].tolist()


richness_importance["predictor"] = pd.Categorical(
    richness_importance["predictor"],
    categories=order,
    ordered=True
)

endemism_importance["predictor"] = pd.Categorical(
    endemism_importance["predictor"],
    categories=order,
    ordered=True
)

richness_importance = richness_importance.sort_values(
    "predictor"
)

endemism_importance = endemism_importance.sort_values(
    "predictor"
)


# =========================================================
# FIGURE
# =========================================================

fig, axes = plt.subplots(
    1, 2,
    figsize=(16, 8),
    sharey=True
)


# =========================================================
# SPECIES RICHNESS
# =========================================================

axes[0].barh(
    richness_importance["label"],
    richness_importance["importance"],
    xerr=richness_importance["std"],
    capsize=3
)

axes[0].set_title(
    "Species richness",
    fontsize=15,
    fontweight="bold"
)

axes[0].set_xlabel(
    "Increase in RMSE after permutation",
    fontsize=12,
    fontweight="bold"
)

axes[0].set_ylabel(
    "Predictor",
    fontsize=12,
    fontweight="bold"
)


# =========================================================
# SPECIES ENDEMISM
# =========================================================

axes[1].barh(
    endemism_importance["label"],
    endemism_importance["importance"],
    xerr=endemism_importance["std"],
    capsize=3
)

axes[1].set_title(
    "Species endemism",
    fontsize=15,
    fontweight="bold"
)

axes[1].set_xlabel(
    "Increase in RMSE after permutation",
    fontsize=12,
    fontweight="bold"
)

axes[1].set_ylabel("")


# =========================================================
# FORMATTING
# =========================================================

for ax in axes:

    ax.tick_params(
        axis="both",
        labelsize=10
    )

    ax.grid(
        axis="x",
        alpha=0.3
    )


fig.suptitle(
    "Predictor importance for biodiversity",
    fontsize=18,
    fontweight="bold"
)

plt.tight_layout()

plt.savefig(
    "predictor_importance_richness_endemism.png",
    dpi=600,
    bbox_inches="tight"
)

plt.show()
```

INFLUENCE OF EACH PREDICTORS CATEGORIES

```python
from collections import defaultdict

# Organise predictors by category.
predictor_groups = defaultdict(list)

for variable, (group, label) in predictor_info.items():
    predictor_groups[group].append(variable)

print("Predictor groups:")

for group, variables in predictor_groups.items():
    print(f"{group}: {len(variables)} predictors")
    print(variables)
```

```python
# Models must already be fitted.
models = {"Species richness": richness_models['mosaic_vegetation_richness'],"Endemism": endemism_models['mosaic_vegetation_richness']}

# Retain predictors available in the grid.
all_predictors = [variable for variable in predictor_info if variable in df_model.columns]

missing_predictors = [variable for variable in predictor_info if variable not in df_model.columns]

if missing_predictors:
    print("Predictors missing from grid:")
    print(missing_predictors)

if not all_predictors:
    raise ValueError("No predictor columns were found in grid_gdf.")

# Keep only complete predictor records for this initial analysis.
# In production, handle missing values using the same procedure
# used during model fitting.
analysis_gdf = df_model.loc[
    df_model[all_predictors].notna().all(axis=1)
].copy()

X = analysis_gdf[all_predictors].copy()

print("Grid cells used:", len(analysis_gdf))
print("Predictors used:", len(all_predictors))
```

```python
prediction_columns = {}

for response, model in models.items():

    pred = np.asarray(model.predict(X)).reshape(-1)

    prediction_columns[response] = pred

    analysis_gdf[f"predicted_{response.lower().replace(' ', '_')}"] = pred

print("Prediction summaries:")

for response, pred in prediction_columns.items():
    print(f"\n{response}")
    print(pd.Series(pred).describe())
```

```python
fig, axes = plt.subplots(
    1, 2,
    figsize=(16, 8),
    layout="constrained"
)

map_variables = [
    ("predicted_species_richness", "Predicted species richness"),
    ("predicted_endemism", "Predicted endemism")
]

for ax, (column, title) in zip(axes, map_variables):

    analysis_gdf.plot(
        column=column,
        ax=ax,
        legend=True,
        cmap="viridis",
        missing_kwds={"color": "lightgrey"}
    )

    ax.set_title(title, fontsize=14)
    ax.set_axis_off()

plt.show()
```

```python
# Number of repeated permutations
N_PERMUTATIONS = 30

rng = np.random.default_rng(42)

# Store prediction changes by response and category.
effect_records = []

# Store baseline predictions for reference.
baseline_predictions = {}

for response, model in models.items():

    baseline = np.asarray(model.predict(X)).reshape(-1)

    baseline_predictions[response] = baseline

    for group, variables in predictor_groups.items():

        # Only use predictors included in the model and grid.
        group_variables = [
            variable
            for variable in variables
            if variable in X.columns
        ]

        if not group_variables:
            continue

        # Repeat permutations.
        changes = []

        for repeat in range(N_PERMUTATIONS):

            X_permuted = X.copy()

            # Apply the SAME permutation to all variables
            # in this category, preserving their internal
            # row-wise relationships.
            permutation = rng.permutation(len(X))

            X_permuted.loc[:, group_variables] = (
                X.iloc[permutation][group_variables].to_numpy()
            )

            permuted_predictions = np.asarray(
                model.predict(X_permuted)
            ).reshape(-1)

            # Absolute prediction change per grid cell.
            delta = np.abs(
                baseline - permuted_predictions
            )

            changes.append(delta)

        changes = np.vstack(changes)

        # Average sensitivity for each grid cell.
        mean_change = changes.mean(axis=0)

        # Retain all grid-cell values for boxplots.
        for value in mean_change:

            effect_records.append({
                "response": response,
                "group": group,
                "prediction_change": value
            })

effect_df = pd.DataFrame(effect_records)

print(effect_df.head())
```

```python
# Category order
category_order = [
    group for group in predictor_groups
    if group in effect_df["group"].unique()
]

fig, axes = plt.subplots(
    1, 2,
    figsize=(17, 7),
    layout="constrained"
)

responses = ["Species richness", "Endemism"]

for ax, response in zip(axes, responses):

    subset = effect_df[
        effect_df["response"] == response
    ]

    box_data = [
        subset.loc[
            subset["group"] == group,
            "prediction_change"
        ].to_numpy()
        for group in category_order
    ]

    ax.boxplot(
        box_data,
        tick_labels=category_order,
        showfliers=False
    )

    ax.set_title(
        f"Predictor-group sensitivity: {response}",
        fontsize=13
    )

    ax.set_ylabel(
        "Absolute change in predicted value"
    )

    ax.set_xlabel("Predictor category")

    ax.tick_params(
        axis="x",
        rotation=45
    )

    ax.grid(
        axis="y",
        alpha=0.3
    )

plt.show()
```

```python
importance_summary = (
    effect_df
    .groupby(["response", "group"])["prediction_change"]
    .agg(["mean", "median", "std"])
    .reset_index()
    .sort_values(["response", "mean"], ascending=[True, False])
)

print(importance_summary)
```

```python
fig, axes = plt.subplots(
    1, 2,
    figsize=(15, 6),
    layout="constrained"
)

for ax, response in zip(axes, responses):

    subset = importance_summary[
        importance_summary["response"] == response
    ].sort_values("mean")

    ax.barh(
        subset["group"],
        subset["mean"]
    )

    ax.set_title(
        f"Mean prediction sensitivity: {response}"
    )

    ax.set_xlabel(
        "Mean absolute prediction change"
    )

    ax.set_ylabel("Predictor category")

plt.show()
```

```python
# Define a different colour for each predictor group
group_colors = {
    "river": "#1f77b4",
    "vegetation": "#2ca02c",
    "terrain": "#ff7f0e",
    "refugia": "#9467bd",
    "Paleoclimate": "#d62728",
    "Forest": "#8c564b",
    "Soil": "#17becf"
}

# Order groups by median prediction change
group_order = (
    effect_df.groupby("group")["prediction_change"]
    .median()
    .sort_values()
    .index.tolist()
)

responses = ["Species richness", "Endemism"]

fig, axes = plt.subplots(
    1, 2,
    figsize=(16, 8),
    layout="constrained"
)

for ax, response in zip(axes, responses):

    subset = effect_df[effect_df["response"] == response]

    # Prepare data for each group
    box_data = [
        subset.loc[
            subset["group"] == group,
            "prediction_change"
        ].dropna().values
        for group in group_order
    ]

    # Create boxplots
    bp = ax.boxplot(
        box_data,
        vert=False,
        tick_labels=group_order,
        showfliers=False,
        patch_artist=True,
        medianprops={
            "color": "black",
            "linewidth": 1.5
        },
        whiskerprops={"linewidth": 1.2},
        capprops={"linewidth": 1.2},
        boxprops={"linewidth": 1.2}
    )

    # Assign a distinct colour to each predictor group
    for box, group in zip(bp["boxes"], group_order):
        box.set_facecolor(group_colors.get(group, "lightgray"))
        box.set_alpha(0.75)

    ax.set_title(
        response,
        fontsize=14,
        fontweight="bold"
    )

    ax.set_xlabel("Absolute change in predicted value", fontsize=11)
    ax.set_ylabel("Predictor group", fontsize=11)

    ax.grid(
        axis="x",
        linestyle="--",
        alpha=0.3
    )

fig.suptitle(
    "Sensitivity of biodiversity predictions to predictor groups",
    fontsize=16,
    fontweight="bold"
)

plt.tight_layout()

plt.savefig(
    "sensitivity_biodiversity_predictions_predictor_groups.png",
    dpi=600,
    bbox_inches="tight"
)
plt.show()
```

```python
from statsmodels.discrete.discrete_model import NegativeBinomial

mammal_richness_models = {}

df_model_mammal["ORD_STRA"] = mammals["ORD_STRA"].astype("category") # Make sure this ORDER_STRA is treaten as categorical

for vegetation_var in vegetation_vars:

    formula = (
        "richness ~ "
        "C(ORD_STRA) + "
        "distance_km + "
        "LENGTH_KM + "
        "DIS_AV_CMS + "
        f"{vegetation_var} + "
        "slope_std + "
        "TPI_mean + "
        "TRI_mean +"
        "TRI_std +"
        "roughness_std + "
        "hillshade_mean + "
        "hillshade_std + "
        "refugia_dist_km + "
        "BIO1_temporal_mean + "
        "BIO1_temporal_SD + "
        "BIO12_temporal_mean + "
        "BIO12_temporal_SD + "
        "forest_loss_mean + "
        "forest_canopy_mean + "
        "forest_canopy_stdev + "
        "tree_cover_mean + "
        "tree_cover_stdev + "
        "clay_mean + "
        "clay_stdev + "
        "cec_mean + "
        "cec_stdev + "
        "soc_mean + "
        "soc_stdev + "
        "ph_mean + "
        "ph_stdev"
    )

    model_mammal_richness = NegativeBinomial.from_formula(formula,data=df_model_mammal).fit(method="bfgs",maxiter=2000,disp=True)

    mammal_richness_models[vegetation_var] = model_mammal_richness

    print("\n")
    print("=" * 80)
    print(vegetation_var)
    print("=" * 80)
    print(model_mammal_richness.summary())
```

```python
mammal_richness_comparison = pd.DataFrame({
    "mammal_model": [
        "Vegetation richness",
        "Vegetation Shannon",
        "Vegetation evenness"
    ],
    "AIC": [
        mammal_richness_models["mosaic_vegetation_richness"].aic,
        mammal_richness_models["mosaic_vegetation_shannon"].aic,
        mammal_richness_models["mosaic_vegetation_evenness"].aic
    ]
})

print(mammal_richness_comparison)
```

```python
import statsmodels.api as sm
import statsmodels.formula.api as smf

mammal_endemism_models = {}

df_model_mammal["ORD_STRA"] = grid["ORD_STRA"].astype("category") # Make sure this ORDER_STRA is treaten as categorical

for vegetation_var in vegetation_vars:

    formula = (
        "WE ~ "
        "C(ORD_STRA) + "
        "distance_km + "
        "LENGTH_KM + "
        "DIS_AV_CMS + "
        f"{vegetation_var} + "
        "slope_std + "
        "TPI_mean + "
        "TRI_mean +"
        "TRI_std +"
        "roughness_std + "
        "hillshade_mean + "
        "hillshade_std + "
        "refugia_dist_km + "
        "BIO1_temporal_mean + "
        "BIO1_temporal_SD + "
        "BIO12_temporal_mean + "
        "BIO12_temporal_SD + "
        "forest_loss_mean + "
        "forest_canopy_mean + "
        "forest_canopy_stdev + "
        "tree_cover_mean + "
        "tree_cover_stdev + "
        "clay_mean + "
        "clay_stdev + "
        "cec_mean + "
        "cec_stdev + "
        "soc_mean + "
        "soc_stdev + "
        "ph_mean + "
        "ph_stdev"
    )

    model_mammal_endemism = smf.glm(formula=formula,data=df_model_mammal,family=sm.families.Gamma(link=sm.families.links.Log())).fit()
    mammal_endemism_models[vegetation_var] = model_mammal_endemism

    print("\n")
    print("=" * 80)
    print(vegetation_var)
    print("=" * 80)
    print(model_mammal_endemism.summary())
```

```python
import numpy as np
import pandas as pd
import geopandas as gpd
import matplotlib.pyplot as plt
from matplotlib.colors import TwoSlopeNorm
from matplotlib.cm import ScalarMappable

df_model_mammal[['grid','geometry']] = mammals[['grid','geometry']]
df_model_mammal["richness_pred"] = mammal_richness_models['mosaic_vegetation_richness'].predict(df_model_mammal)
df_model_mammal["endemism_pred"] = mammal_endemism_models['mosaic_vegetation_richness'].predict(df_model_mammal)
df_model_mammal["richness_residual"] = (df_model_mammal["richness"] - df_model_mammal["richness_pred"])
df_model_mammal["endemism_residual"] = (df_model_mammal["WE"] - df_model_mammal["endemism_pred"])
df_model_mammal = gpd.GeoDataFrame(df_model_mammal,geometry="geometry",crs="EPSG:4326")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import Normalize, TwoSlopeNorm
from matplotlib.cm import ScalarMappable


fig, axes = plt.subplots(
    2, 3,
    figsize=(18, 11)
)


# =========================================================
# DATA
# =========================================================

responses = [
    {
        "name": "Mammals richness",
        "observed": "richness",
        "predicted": "richness_pred",
        "residual": "richness_residual"
    },
    {
        "name": "Mammals endemism",
        "observed": "WE",
        "predicted": "endemism_pred",
        "residual": "endemism_residual"
    }
]


# =========================================================
# LOOP
# =========================================================

for row, response in enumerate(responses):

    observed = response["observed"]
    predicted = response["predicted"]
    residual = response["residual"]


    # =====================================================
    # OBSERVED + PREDICTED SCALE
    # =====================================================

    combined = np.concatenate([
        df_model_mammal[observed].dropna().values,
        df_model_mammal[predicted].dropna().values
    ])

    vmin = np.percentile(combined, 1)
    vmax = np.percentile(combined, 99)


    # =====================================================
    # RESIDUAL SCALE
    # =====================================================

    residual_values = df_model_mammal[residual].dropna().values

    residual_max = np.percentile(
        np.abs(residual_values),
        99
    )


    # =====================================================
    # OBSERVED
    # =====================================================

    df_model_mammal.plot(
        column=observed,
        ax=axes[row, 0],
        cmap="viridis",
        vmin=vmin,
        vmax=vmax,
        linewidth=0,
        edgecolor="none"
    )

    axes[row, 0].set_title(
        "Observed",
        fontsize=14,
        fontweight="bold"
    )


    # =====================================================
    # PREDICTED
    # =====================================================

    df_model_mammal.plot(
        column=predicted,
        ax=axes[row, 1],
        cmap="viridis",
        vmin=vmin,
        vmax=vmax,
        linewidth=0,
        edgecolor="none"
    )

    axes[row, 1].set_title(
        "Predicted",
        fontsize=14,
        fontweight="bold"
    )


    # =====================================================
    # RESIDUAL
    # =====================================================

    df_model_mammal.plot(
        column=residual,
        ax=axes[row, 2],
        cmap="RdBu_r",
        vmin=-residual_max,
        vmax=residual_max,
        linewidth=0,
        edgecolor="none"
    )

    axes[row, 2].set_title(
        "Residual",
        fontsize=14,
        fontweight="bold"
    )


    # =====================================================
    # ROW LABEL
    # =====================================================

    axes[row, 0].set_ylabel(
        response["name"],
        fontsize=14,
        fontweight="bold"
    )


    # =====================================================
    # REMOVE AXES
    # =====================================================

    for col in range(3):
        axes[row, col].set_axis_off()


    # =====================================================
    # OBSERVED/PREDICTED COLORBAR
    # =====================================================

    sm = ScalarMappable(
        norm=Normalize(
            vmin=vmin,
            vmax=vmax
        ),
        cmap="viridis"
    )

    sm.set_array([])

    cbar = fig.colorbar(
        sm,
        ax=axes[row, 0:2],
        fraction=0.025,
        pad=0.02
    )

    cbar.set_label(
        response["name"],
        fontsize=11,
        fontweight="bold"
    )


    # =====================================================
    # RESIDUAL COLORBAR
    # =====================================================

    sm_res = ScalarMappable(
        norm=TwoSlopeNorm(
            vmin=-residual_max,
            vcenter=0,
            vmax=residual_max
        ),
        cmap="RdBu_r"
    )

    sm_res.set_array([])

    cbar_res = fig.colorbar(
        sm_res,
        ax=axes[row, 2],
        fraction=0.025,
        pad=0.02
    )

    cbar_res.set_label(
        "Pearson residual",
        fontsize=11,
        fontweight="bold"
    )


# =========================================================
# TITLE
# =========================================================

fig.suptitle(
    "Observed, predicted and residual Mammal biodiversity",
    fontsize=18,
    fontweight="bold",
    y=0.98
)

plt.tight_layout()

plt.savefig(
    "observed_predicted_residual_mammal_biodiversity.png",
    dpi=600,
    bbox_inches="tight"
)

plt.show()
```

 # Bird modelling 

```python
from sklearn.preprocessing import StandardScaler

predictors = (river_continous_vars + vegetation_vars + terrain_vars + refugia_vars + paleoclimate_vars + forest_vars + soil_vars)

df_model_bird = birds[["richness", "WE", "CWE"] + predictors].copy()

# Complete cases so all six models use exactly the same observations
df_model_bird = df_model_bird.dropna().copy()

# Standardize predictors
scaler = StandardScaler()

df_model_bird[predictors] = scaler.fit_transform(
    df_model_bird[predictors]
)
```

## Bird Richness

```python
mean_richness = df_model_bird["richness"].mean()
var_richness = df_model_bird["richness"].var()

print("Mean:", mean_richness)
print("Variance:", var_richness)
print("Variance / Mean:", var_richness / mean_richness)
```

```python
from statsmodels.discrete.discrete_model import NegativeBinomial

bird_richness_models = {}

df_model_bird["ORD_STRA"] = birds["ORD_STRA"].astype("category") # Make sure this ORDER_STRA is treaten as categorical

for vegetation_var in vegetation_vars:

    formula = (
        "richness ~ "
        "C(ORD_STRA) + "
        "distance_km + "
        "LENGTH_KM + "
        "DIS_AV_CMS + "
        f"{vegetation_var} + "
        "slope_std + "
        "TPI_mean + "
        "TRI_mean +"
        "TRI_std +"
        "roughness_std + "
        "hillshade_mean + "
        "hillshade_std + "
        "refugia_dist_km + "
        "BIO1_temporal_mean + "
        "BIO1_temporal_SD + "
        "BIO12_temporal_mean + "
        "BIO12_temporal_SD + "
        "forest_loss_mean + "
        "forest_canopy_mean + "
        "forest_canopy_stdev + "
        "tree_cover_mean + "
        "tree_cover_stdev + "
        "clay_mean + "
        "clay_stdev + "
        "cec_mean + "
        "cec_stdev + "
        "soc_mean + "
        "soc_stdev + "
        "ph_mean + "
        "ph_stdev"
    )

    model_bird_richness = NegativeBinomial.from_formula(formula,data=df_model_bird).fit(method="bfgs",maxiter=2000,disp=True)

    bird_richness_models[vegetation_var] = model_bird_richness

    print("\n")
    print("=" * 80)
    print(vegetation_var)
    print("=" * 80)
    print(model_bird_richness.summary())
```

```python
import statsmodels.api as sm
import statsmodels.formula.api as smf

bird_endemism_models = {}

df_model_bird["ORD_STRA"] = birds["ORD_STRA"].astype("category") # Make sure this ORDER_STRA is treaten as categorical

for vegetation_var in vegetation_vars:

    formula = (
        "WE ~ "
        "C(ORD_STRA) + "
        "distance_km + "
        "LENGTH_KM + "
        "DIS_AV_CMS + "
        f"{vegetation_var} + "
        "slope_std + "
        "TPI_mean + "
        "TRI_mean +"
        "TRI_std +"
        "roughness_std + "
        "hillshade_mean + "
        "hillshade_std + "
        "refugia_dist_km + "
        "BIO1_temporal_mean + "
        "BIO1_temporal_SD + "
        "BIO12_temporal_mean + "
        "BIO12_temporal_SD + "
        "forest_loss_mean + "
        "forest_canopy_mean + "
        "forest_canopy_stdev + "
        "tree_cover_mean + "
        "tree_cover_stdev + "
        "clay_mean + "
        "clay_stdev + "
        "cec_mean + "
        "cec_stdev + "
        "soc_mean + "
        "soc_stdev + "
        "ph_mean + "
        "ph_stdev "
    )

    model_bird_endemism = smf.glm(formula=formula,data=df_model_bird,family=sm.families.Gamma(link=sm.families.links.Log())).fit(method="lbfgs")
    bird_endemism_models[vegetation_var] = model_bird_endemism

    print("\n")
    print("=" * 80)
    print(vegetation_var)
    print("=" * 80)
    print(model_bird_endemism.summary())
```

```python
fig, axes = plt.subplots(
    2, 3,
    figsize=(18, 11)
)


# =========================================================
# DATA
# =========================================================

responses = [
    {
        "name": "Amphibians richness",
        "observed": "richness",
        "predicted": "richness_pred",
        "residual": "richness_residual"
    },
    {
        "name": "Amphibians endemism",
        "observed": "WE",
        "predicted": "endemism_pred",
        "residual": "endemism_residual"
    }
]


# =========================================================
# LOOP
# =========================================================

for row, response in enumerate(responses):

    observed = response["observed"]
    predicted = response["predicted"]
    residual = response["residual"]


    # =====================================================
    # OBSERVED + PREDICTED SCALE
    # =====================================================

    combined = np.concatenate([
        df_model_amphibian[observed].dropna().values,
        df_model_amphibian[predicted].dropna().values
    ])

    vmin = np.percentile(combined, 1)
    vmax = np.percentile(combined, 99)


    # =====================================================
    # RESIDUAL SCALE
    # =====================================================

    residual_values = df_model_amphibian[residual].dropna().values

    residual_max = np.percentile(
        np.abs(residual_values),
        99
    )


    # =====================================================
    # OBSERVED
    # =====================================================

    df_model_amphibian.plot(
        column=observed,
        ax=axes[row, 0],
        cmap="viridis",
        vmin=vmin,
        vmax=vmax,
        linewidth=0,
        edgecolor="none"
    )

    axes[row, 0].set_title(
        "Observed",
        fontsize=14,
        fontweight="bold"
    )


    # =====================================================
    # PREDICTED
    # =====================================================

    df_model_amphibian.plot(
        column=predicted,
        ax=axes[row, 1],
        cmap="viridis",
        vmin=vmin,
        vmax=vmax,
        linewidth=0,
        edgecolor="none"
    )

    axes[row, 1].set_title(
        "Predicted",
        fontsize=14,
        fontweight="bold"
    )


    # =====================================================
    # RESIDUAL
    # =====================================================

    df_model_amphibian.plot(
        column=residual,
        ax=axes[row, 2],
        cmap="RdBu_r",
        vmin=-residual_max,
        vmax=residual_max,
        linewidth=0,
        edgecolor="none"
    )

    axes[row, 2].set_title(
        "Residual",
        fontsize=14,
        fontweight="bold"
    )


    # =====================================================
    # ROW LABEL
    # =====================================================

    axes[row, 0].set_ylabel(
        response["name"],
        fontsize=14,
        fontweight="bold"
    )


    # =====================================================
    # REMOVE AXES
    # =====================================================

    for col in range(3):
        axes[row, col].set_axis_off()


    # =====================================================
    # OBSERVED/PREDICTED COLORBAR
    # =====================================================

    sm = ScalarMappable(
        norm=Normalize(
            vmin=vmin,
            vmax=vmax
        ),
        cmap="viridis"
    )

    sm.set_array([])

    cbar = fig.colorbar(
        sm,
        ax=axes[row, 0:2],
        fraction=0.025,
        pad=0.02
    )

    cbar.set_label(
        response["name"],
        fontsize=11,
        fontweight="bold"
    )


    # =====================================================
    # RESIDUAL COLORBAR
    # =====================================================

    sm_res = ScalarMappable(
        norm=TwoSlopeNorm(
            vmin=-residual_max,
            vcenter=0,
            vmax=residual_max
        ),
        cmap="RdBu_r"
    )

    sm_res.set_array([])

    cbar_res = fig.colorbar(
        sm_res,
        ax=axes[row, 2],
        fraction=0.025,
        pad=0.02
    )

    cbar_res.set_label(
        "Pearson residual",
        fontsize=11,
        fontweight="bold"
    )


# =========================================================
# TITLE
# =========================================================

fig.suptitle(
    "Observed, predicted and residual Amphibians biodiversity",
    fontsize=18,
    fontweight="bold",
    y=0.98
)

plt.tight_layout()

plt.savefig(
    "observed_predicted_residual_amphibian_biodiversity.png",
    dpi=600,
    bbox_inches="tight"
)

plt.show()
```

```python
import pandas as pd
import numpy as np


def extract_glm_results(models):

    results = []

    for response, taxa_models in models.items():

        for taxon, model in taxa_models.items():

            # Extract coefficients
            params = model.params

            # Standard errors
            se = model.bse

            # P-values
            pvalues = model.pvalues

            # Confidence intervals
            ci = model.conf_int()

            for predictor in params.index:

                beta = params[predictor]

                lower = ci.loc[predictor, 0]

                upper = ci.loc[predictor, 1]

                results.append({

                    "response": response,

                    "taxon": taxon,

                    "predictor": predictor,
                    
                     "Predictor_label": predictor_labels.get(
                        predictor, predictor
                    ),

                    "estimate": beta,

                    "std_error": se[predictor],

                    "z_value": model.tvalues[predictor],

                    "p_value": pvalues[predictor],

                    "lower": lower,

                    "upper": upper,

                    "exp_estimate": np.exp(beta),

                    "exp_lower": np.exp(lower),

                    "exp_upper": np.exp(upper),

                    "n_observations": int(model.nobs),

                    "aic": model.aic

                })

    return pd.DataFrame(results)


# Extract all models

model_results = extract_glm_results(models)


# Display results

display(model_results.head())


# Export

model_results.to_csv("data/model_effects.csv",index=False)
```

```python
import numpy as np
import pandas as pd
from scipy.stats import chi2


def calculate_grouped_importance(model, predictor_groups):

    params = model.params
    covariance = model.cov_params()

    results = []

    for category, variables in predictor_groups.items():

        # Identify model coefficients belonging to this category
        selected_terms = []

        for term in params.index:

            for variable in variables:

                # Handles categorical ORD_STRA contrasts
                if (
                    term == variable
                    or term.startswith(f"C({variable})")
                    or term.startswith(f"{variable}[")
                ):
                    selected_terms.append(term)
                    break

        if len(selected_terms) == 0:
            continue

        beta = params.loc[selected_terms].to_numpy()

        cov = covariance.loc[
            selected_terms,
            selected_terms
        ].to_numpy()

        # Joint Wald statistic
        wald_stat = float(
            beta.T @ np.linalg.pinv(cov) @ beta
        )

        degrees_freedom = len(selected_terms)

        p_value = chi2.sf(
            wald_stat,
            degrees_freedom
        )

        results.append({
            "Category": category,
            "Wald_statistic": wald_stat,
            "df": degrees_freedom,
            "P_value": p_value
        })

    importance = pd.DataFrame(results)

    # Normalize Wald statistics into percentages
    importance["Importance"] = (
        importance["Wald_statistic"]
        / importance["Wald_statistic"].sum()
    ) * 100

    return importance
```

```python
importance_results = []

for response, taxon_models in models.items():

    for taxon, model in taxon_models.items():

        result = calculate_grouped_importance(
            model,
            predictor_groups
        )

        result["Response"] = response
        result["Taxon"] = taxon

        importance_results.append(result)


importance_df = pd.concat(
    importance_results,
    ignore_index=True
)

importance_df.to_csv(
    "Congo_Basin_grouped_importance.csv",
    index=False
)

print(importance_df)
```

```python
def extract_coefficients(models):

    results = []

    for response, taxon_models in models.items():

        for taxon, model in taxon_models.items():

            params = model.params
            conf = model.conf_int()

            for term in params.index:

                if term in ["Intercept", "alpha"]:
                    continue

                results.append({

                    "Response": response,
                    "Taxon": taxon,
                    "Predictor": term,
                    
                    "Predictor_label": predictor_labels.get(term, term),
                    
                    "Coefficient": params[term],

                    "CI_lower": conf.loc[term, 0],

                    "CI_upper": conf.loc[term, 1],

                    "P_value": model.pvalues[term]

                })

    return pd.DataFrame(results)


coefficient_df = extract_coefficients(models)
```

```python
import re
import copy
import warnings
import numpy as np
import pandas as pd
import patsy
import statsmodels.formula.api as smf

from scipy.stats import chi2


def remove_predictor_group(formula, variables):

    description = patsy.ModelDesc.from_formula(formula)

    retained_terms = []

    for term in description.rhs_termlist:

        term_name = term.name()

        # Always retain the intercept
        if term_name == "Intercept":
            retained_terms.append(term)
            continue

        # Remove any term involving a predictor in this group.
        # This also removes interactions involving that predictor.
        remove_term = False

        for variable in variables:

            pattern = (
                r"(?<![A-Za-z0-9_])"
                + re.escape(variable)
                + r"(?![A-Za-z0-9_])"
            )

            if re.search(pattern, term_name):
                remove_term = True
                break

        if not remove_term:
            retained_terms.append(term)

    reduced_description = patsy.ModelDesc(
        description.lhs_termlist,
        retained_terms
    )

    return reduced_description.describe()


def refit_reduced_model(full_model, reduced_formula, data):

    model_class = full_model.model.__class__.__name__

    # Discrete Negative Binomial
    if model_class == "NegativeBinomial":

        reduced_model = smf.negativebinomial(
            formula=reduced_formula,
            data=data
        ).fit(
            disp=0,
            maxiter=300
        )

    # GLM, including Gamma
    elif model_class == "GLM":

        family = copy.deepcopy(
            full_model.model.family
        )

        reduced_model = smf.glm(
            formula=reduced_formula,
            data=data,
            family=family
        ).fit()

    else:
        raise TypeError(
            f"Unsupported model class: {model_class}. "
            "Check the fitted model type."
        )

    return reduced_model


def grouped_likelihood_ratio_test(
    full_model,
    predictor_groups
):

    formula = full_model.model.formula

    # Retrieve the original fitting data
    original_data = full_model.model.data.frame.copy()

    # Match the exact observations used in the full model
    row_labels = full_model.model.data.row_labels

    if row_labels is not None:
        data = original_data.loc[row_labels].copy()
    else:
        data = original_data.copy()

    full_llf = full_model.llf
    full_aic = full_model.aic
    full_df = full_model.df_model

    results = []

    for category, variables in predictor_groups.items():

        reduced_formula = remove_predictor_group(
            formula,
            variables
        )

        if reduced_formula == formula:
            continue

        reduced_model = refit_reduced_model(
            full_model,
            reduced_formula,
            data
        )

        # Likelihood-ratio statistic
        lr_stat = 2 * (
            full_llf - reduced_model.llf
        )

        # Difference in model degrees of freedom
        df_difference = int(
            round(full_df - reduced_model.df_model)
        )

        if df_difference <= 0:
            warnings.warn(
                f"No positive degrees-of-freedom difference "
                f"for {category}. Check the formula."
            )
            continue

        if lr_stat < -1e-5:
            warnings.warn(
                f"Negative LR statistic for {category}. "
                "Check convergence and model fitting."
            )

        lr_stat = max(0, lr_stat)

        p_value = chi2.sf(
            lr_stat,
            df_difference
        )

        results.append({

            "Category": category,

            "LR_statistic": lr_stat,

            "df": df_difference,

            "P_value": p_value,

            "Delta_AIC": reduced_model.aic - full_aic,

            "Full_logLik": full_llf,

            "Reduced_logLik": reduced_model.llf,

            "Reduced_formula": reduced_formula,

            "Converged": getattr(
                reduced_model,
                "converged",
                True
            )

        })

    return pd.DataFrame(results)
```

```python
lrt_results = []

for response, taxon_models in models.items():

    for taxon, fitted_model in taxon_models.items():

        print(f"Testing {response} - {taxon}")

        result = grouped_likelihood_ratio_test(
            full_model=fitted_model,
            predictor_groups=predictor_groups
        )

        result["Response"] = response
        result["Taxon"] = taxon

        lrt_results.append(result)


lrt_df = pd.concat(
    lrt_results,
    ignore_index=True
)

lrt_df.to_csv(
    "Congo_Basin_grouped_LRT_results.csv",
    index=False
)

print(
    lrt_df[
        [
            "Response",
            "Taxon",
            "Category",
            "LR_statistic",
            "df",
            "P_value",
            "Delta_AIC"
        ]
    ].round(4)
)
```

```python

import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd

from matplotlib.colors import Normalize
from matplotlib.cm import ScalarMappable

# ============================================================
# 1. GENERAL VISUAL SETTINGS
# ============================================================

sns.set_theme(
    style="white",
    context="paper",
    font="DejaVu Sans",
    font_scale=1.15
)

plt.rcParams.update({
    "font.family": "DejaVu Sans",
    "font.size": 11,
    "axes.titlesize": 14,
    "axes.labelsize": 11,
    "xtick.labelsize": 10,
    "ytick.labelsize": 11,
    "figure.titlesize": 19,
    "figure.dpi": 150,
    "savefig.dpi": 600,
    "axes.spines.top": False,
    "axes.spines.right": False
})

# ============================================================
# 2. CATEGORY AND TAXON LABELS
# ============================================================

category_order = list(predictor_groups.keys())

category_labels = {
    "Hydrology": "Hydrology",
    "Vegetation": "Vegetation",
    "Topography": "Topography",
    "Climate": "Climate",
    "Palaeogeography": "Palaeogeography",
    "Forest structure": "Forest structure",
    "Anthropogenic pressure": "Anthropogenic\npressure",
    "Soil": "Soil properties"
}

taxon_labels = {
    "Mammals": "Mammals",
    "Birds": "Birds",
    "Amphibians": "Amphibians",
    "Reptiles": "Reptiles",
    "All taxa": "All taxa"
}

# ============================================================
# 3. SIGNIFICANCE CLASSIFICATION
# ============================================================

def significance_label(p):

    if pd.isna(p):
        return ""

    if p < 0.001:
        return "***"

    elif p < 0.01:
        return "**"

    elif p < 0.05:
        return "*"

    else:
        return "ns"


def format_annotation(lr, p):

    if pd.isna(lr):
        return ""

    return f"{lr:.1f}\n{significance_label(p)}"


# ============================================================
# 4. PREPARE DATA
# ============================================================

plot_data = lrt_df.copy()

plot_data["Taxon"] = pd.Categorical(
    plot_data["Taxon"],
    categories=taxon_order,
    ordered=True
)

plot_data["Category"] = pd.Categorical(
    plot_data["Category"],
    categories=category_order,
    ordered=True
)

# Shared colour scale across both responses
vmin = 0

vmax = plot_data["LR_statistic"].max()

# Avoid a zero-width colour scale
if pd.isna(vmax) or vmax <= 0:
    vmax = 1

# ============================================================
# 5. HEATMAP FUNCTION
# ============================================================

def plot_lrt_heatmap(ax, response, panel_title):

    subset = plot_data[
        plot_data["Response"] == response
    ].copy()

    # Matrix containing LRT statistics
    matrix = (
        subset.pivot(
            index="Taxon",
            columns="Category",
            values="LR_statistic"
        )
        .reindex(
            index=taxon_order,
            columns=category_order
        )
    )

    # Annotation matrix
    annotation = pd.DataFrame(
        "",
        index=matrix.index,
        columns=matrix.columns
    )

    for _, row in subset.iterrows():

        taxon = row["Taxon"]
        category = row["Category"]

        annotation.loc[taxon, category] = format_annotation(
            row["LR_statistic"],
            row["P_value"]
        )

    # Heatmap
    sns.heatmap(
        matrix,
        ax=ax,
        cmap="YlOrRd",
        vmin=vmin,
        vmax=vmax,
        annot=annotation,
        fmt="",
        annot_kws={
            "fontsize": 10,
            "fontweight": "medium",
            "ha": "center",
            "va": "center"
        },
        linewidths=1.2,
        linecolor="white",
        square=False,
        cbar=False,
        mask=matrix.isna()
    )

    # Titles and labels
    ax.set_title(
        panel_title,
        loc="left",
        fontsize=14,
        fontweight="bold",
        pad=15
    )

    ax.set_xlabel(
        "Ecological predictor category",
        fontsize=11,
        fontweight="bold",
        labelpad=12
    )

    ax.set_ylabel(
        "Taxonomic group",
        fontsize=11,
        fontweight="bold",
        labelpad=10
    )

    # Axis labels
    ax.set_xticklabels(
        [
            category_labels.get(str(x), str(x))
            for x in matrix.columns
        ],
        rotation=0,
        ha="center"
    )

    ax.set_yticklabels(
        [
            taxon_labels.get(str(y), str(y))
            for y in matrix.index
        ],
        rotation=0
    )

    ax.tick_params(
        axis="both",
        length=0,
        pad=8
    )

    # Clean borders
    for spine in ax.spines.values():
        spine.set_visible(False)


# ============================================================
# 6. CREATE FIGURE
# ============================================================

fig = plt.figure(figsize=(19, 10))

gs = fig.add_gridspec(
    nrows=1,
    ncols=2,
    width_ratios=[1, 1],
    wspace=0.20,
    left=0.07,
    right=0.91,
    top=0.79,
    bottom=0.23
)

ax_A = fig.add_subplot(gs[0, 0])
ax_B = fig.add_subplot(gs[0, 1])

# Richness
plot_lrt_heatmap(
    ax_A,
    response="Richness",
    panel_title="A. Species richness"
)

# Endemism
plot_lrt_heatmap(
    ax_B,
    response="Endemism",
    panel_title="B. Species endemism"
)

# ============================================================
# 7. SHARED COLOUR BAR
# ============================================================

norm = Normalize(vmin=vmin, vmax=vmax)

sm = ScalarMappable(
    norm=norm,
    cmap="YlOrRd"
)

sm.set_array([])

cbar_ax = fig.add_axes([
    0.925,  # left
    0.29,   # bottom
    0.015,  # width
    0.42    # height
])

cbar = fig.colorbar(
    sm,
    cax=cbar_ax
)

cbar.set_label(
    "Likelihood-ratio statistic (LR)",
    fontsize=11,
    fontweight="bold",
    labelpad=12
)

cbar.ax.tick_params(labelsize=10, length=3)

# ============================================================
# 8. FIGURE TITLE
# ============================================================

fig.suptitle(
    "Relative contribution of ecological predictor categories "
    "to Congo Basin biodiversity patterns",
    fontsize=18,
    fontweight="bold",
    y=0.94
)

fig.text(
    0.5,
    0.865,
    "Grouped likelihood-ratio tests from fitted regression models",
    ha="center",
    fontsize=12,
    color="dimgray"
)

# ============================================================
# 9. SIGNIFICANCE LEGEND
# ============================================================

legend_text = (
    "Significance:  *** p < 0.001     "
    "** p < 0.01     "
    "* p < 0.05     "
    "ns p ≥ 0.05"
)

fig.text(
    0.5,
    0.145,
    legend_text,
    ha="center",
    fontsize=10,
    fontweight="medium"
)

# Explanation
fig.text(
    0.5,
    0.085,
    "Values represent likelihood-ratio statistics comparing "
    "the full model with models excluding each predictor category.",
    ha="center",
    fontsize=9,
    color="dimgray"
)

# ============================================================
# 10. SAVE FIGURE
# ============================================================

plt.savefig(
    "Congo_Basin_Grouped_LRT_Heatmaps.png",
    dpi=600,
    bbox_inches="tight",
    facecolor="white"
)

plt.show()
```

```python

import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
import numpy as np
import pandas as pd
import seaborn as sns

# ============================================================
# 1. MODEL OBJECTS
# ============================================================

models = {
    "Carnivores": carnivore_richness_models['mosaic_vegetation_richness'],
    "Herbivores": herbivore_richness_models['mosaic_vegetation_richness'],
    "Omnivores": omnivore_richness_models['mosaic_vegetation_richness']
}

taxon_order = [
    "Carnivores",
    "Herbivores",
    "Omnivores"
]

# ============================================================
# 2. PREDICTOR LABELS
# ============================================================

predictor_labels = {
    "C(ORD_STRA)[T.6]": "River order 6",
    "C(ORD_STRA)[T.7]": "River order 7",
    "C(ORD_STRA)[T.8]": "River order 8",
    "C(ORD_STRA)[T.9]": "River order 9",

    "distance_km": "Distance to river",
    "LENGTH_KM": "River length",
    "DIS_AV_CMS": "Average river discharge",

    "vegetation_richness": "Vegetation richness",
    "vegetation_shannon": "Vegetation Shannon diversity",
    "vegetation_evenness": "Vegetation evenness",

    "slope_std": "Slope variability",
    "TPI_mean": "Mean topographic position index",
    "TRI_mean": "Mean terrain ruggedness index",
    "TRI_std": "Terrain ruggedness variability",
    "roughness_std": "Terrain roughness variability",
    "hillshade_mean": "Mean hillshade",
    "hillshade_std": "Hillshade variability",

    "refugia_dist_km": "Distance to forest refugia",

    "BIO1_temporal_mean": "Mean temperature",
    "BIO1_temporal_SD": "Temperature variability",
    "BIO12_temporal_mean": "Mean precipitation",
    "BIO12_temporal_SD": "Precipitation variability",

    "forest_loss_mean": "Forest loss (2001–2020)",
    "forest_canopy_mean": "Mean canopy height",
    "forest_canopy_stdev": "Canopy height variability",
    "tree_cover_mean": "Mean tree cover",
    "tree_cover_stdev": "Tree cover variability",

    "clay_mean": "Mean soil clay content",
    "clay_stdev": "Soil clay variability",
    "cec_mean": "Mean cation exchange capacity",
    "cec_stdev": "CEC variability",
    "soc_mean": "Mean soil organic carbon",
    "soc_stdev": "Soil organic carbon variability",
    "ph_mean": "Mean soil pH",
    "ph_stdev": "Soil pH variability"
}

# ============================================================
# 3. PREDICTOR CATEGORY CLASSIFICATION
# ============================================================

predictor_groups = {
    "Hydrology": [
        "C(ORD_STRA)[T.6]",
        "C(ORD_STRA)[T.7]",
        "C(ORD_STRA)[T.8]",
        "C(ORD_STRA)[T.9]",
        "distance_km",
        "LENGTH_KM",
        "DIS_AV_CMS"
    ],

    "Vegetation": [
        "vegetation_richness",
        "vegetation_shannon",
        "vegetation_evenness"
    ],

    "Topography": [
        "slope_std",
        "TPI_mean",
        "TRI_mean",
        "TRI_std",
        "roughness_std",
        "hillshade_mean",
        "hillshade_std"
    ],

    "Palaeogeography": [
        "refugia_dist_km"
    ],

    "Paleoclimate": [
        "BIO1_temporal_mean",
        "BIO1_temporal_SD",
        "BIO12_temporal_mean",
        "BIO12_temporal_SD"
    ],

    "Forest structure": [
        "forest_canopy_mean",
        "forest_canopy_stdev",
        "tree_cover_mean",
        "tree_cover_stdev",
        "forest_loss_mean"
    ],



    "Soil": [
        "clay_mean",
        "clay_stdev",
        "cec_mean",
        "cec_stdev",
        "soc_mean",
        "soc_stdev",
        "ph_mean",
        "ph_stdev"
    ]
}

category_colors = {
    "Hydrology": "#2878B5",
    "Vegetation": "#2A9D66",
    "Topography": "#E09F3E",
    "Palaeogeography": "#8064A2",
    "Climate": "#D95F59",
    "Forest structure": "#57A5A0",
    "Soil": "#777777"
}

# ============================================================
# 4. EXTRACT COEFFICIENTS AND CONFIDENCE INTERVALS
# ============================================================

def extract_forest_data(models):

    results = []

    for taxon, model in models.items():

        params = model.params
        conf_int = model.conf_int()

        for predictor in params.index:

            # Exclude intercept
            if predictor.lower() in ["intercept", "const"]:
                continue

            beta = params[predictor]

            ci_low = conf_int.loc[predictor].iloc[0]
            ci_high = conf_int.loc[predictor].iloc[1]

            # Convert coefficients to IRRs
            irr = np.exp(beta)
            irr_low = np.exp(ci_low)
            irr_high = np.exp(ci_high)

            # Identify predictor category
            category = "Other"

            for group, variables in predictor_groups.items():

                if predictor in variables:
                    category = group
                    break

            results.append({
                "Taxon": taxon,
                "Predictor": predictor,
                "Label": predictor_labels.get(
                    predictor, predictor
                ),
                "Category": category,
                "Coefficient": beta,
                "IRR": irr,
                "CI_lower": irr_low,
                "CI_upper": irr_high,
                "Significant": (
                    ci_low > 0 or ci_high < 0
                )
            })

    return pd.DataFrame(results)


forest_df = extract_forest_data(models)

# ============================================================
# 5. DEFINE PREDICTOR ORDER
# ============================================================

predictor_order = [
    predictor
    for group in predictor_groups.values()
    for predictor in group
]

# Retain only predictors present in at least one model
predictor_order = [
    p for p in predictor_order
    if p in forest_df["Predictor"].unique()
]

# Add any predictors not included in the category dictionary
extra_predictors = [
    p for p in forest_df["Predictor"].unique()
    if p not in predictor_order
]

predictor_order += extra_predictors

# ============================================================
# 6. FIGURE STYLE
# ============================================================

sns.set_theme(
    style="white",
    context="paper",
    font="DejaVu Sans"
)

plt.rcParams.update({
    "font.family": "DejaVu Sans",
    "font.size": 10,
    "axes.labelsize": 11,
    "axes.titlesize": 14,
    "xtick.labelsize": 10,
    "ytick.labelsize": 9,
    "figure.titlesize": 18,
    "savefig.dpi": 600
})

# ============================================================
# 7. CREATE FOREST PLOT
# ============================================================

n_predictors = len(predictor_order)

fig_height = max(8, n_predictors * 0.38)

fig, axes = plt.subplots(
    nrows=1,
    ncols=3,
    figsize=(19, fig_height),
    sharey=True
)

# Colours for trophic groups
taxon_colors = {
    "Carnivores": "#B23A48",
    "Herbivores": "#238B57",
    "Omnivores": "#2878B5"
}

# Small vertical offsets for each panel
y_positions = np.arange(n_predictors)

# Determine a shared x-axis range
valid_values = forest_df[
    ["CI_lower", "CI_upper"]
].replace([np.inf, -np.inf], np.nan).dropna()

x_min = max(
    0.01,
    valid_values["CI_lower"].min() * 0.85
)

x_max = valid_values["CI_upper"].max() * 1.15

# ============================================================
# 8. DRAW EACH TROPHIC GROUP
# ============================================================

for ax, taxon in zip(axes, taxon_order):

    subset = forest_df[
        forest_df["Taxon"] == taxon
    ].copy()

    for i, predictor in enumerate(predictor_order):

        row = subset[
            subset["Predictor"] == predictor
        ]

        if row.empty:
            continue

        row = row.iloc[0]

        irr = row["IRR"]
        lower = row["CI_lower"]
        upper = row["CI_upper"]

        category = row["Category"]

        color = category_colors.get(
            category, "#777777"
        )

        # Confidence interval
        ax.plot(
            [lower, upper],
            [i, i],
            color=color,
            linewidth=2.0,
            solid_capstyle="round",
            zorder=2
        )

        # Confidence interval end caps
        ax.plot(
            [lower, lower],
            [i - 0.09, i + 0.09],
            color=color,
            linewidth=1.4
        )

        ax.plot(
            [upper, upper],
            [i - 0.09, i + 0.09],
            color=color,
            linewidth=1.4
        )

        # Point estimate
        ax.scatter(
            irr,
            i,
            color=color,
            s=45,
            edgecolor="white",
            linewidth=0.8,
            zorder=3
        )

    # Reference line: IRR = 1
    ax.axvline(
        1,
        color="#333333",
        linestyle="--",
        linewidth=1.2,
        alpha=0.8,
        zorder=1
    )

    # Axis formatting
    ax.set_xscale("log")
    ax.set_xlim(x_min, x_max)

    ax.set_title(
        taxon,
        fontsize=14,
        fontweight="bold",
        color=taxon_colors[taxon],
        pad=14
    )

    ax.set_xlabel(
        "Incidence rate ratio (IRR, log scale)",
        fontsize=10,
        fontweight="bold",
        labelpad=10
    )

    ax.set_yticks(y_positions)

    ax.set_yticklabels(
        [
            predictor_labels.get(p, p)
            for p in predictor_order
        ]
    )

    ax.invert_yaxis()

    # Subtle grid
    ax.grid(
        axis="x",
        linestyle=":",
        linewidth=0.7,
        alpha=0.5
    )

    ax.grid(axis="y", visible=False)

    ax.tick_params(
        axis="y",
        length=0,
        pad=7
    )

    ax.tick_params(
        axis="x",
        labelsize=9
    )

    # Clean spines
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)
    ax.spines["left"].set_visible(False)

# Only first panel displays predictor labels
for ax in axes[1:]:
    ax.tick_params(
        axis="y",
        labelleft=False
    )

# ============================================================
# 9. MAIN TITLE
# ============================================================

fig.suptitle(
    "Environmental associations with species richness "
    "across trophic groups",
    fontsize=18,
    fontweight="bold",
    y=0.995
)

fig.text(
    0.5,
    0.965,
    "Negative Binomial regression estimates with 95% confidence intervals",
    ha="center",
    fontsize=11,
    color="dimgray"
)

# ============================================================
# 10. ECOLOGICAL CATEGORY LEGEND
# ============================================================

from matplotlib.lines import Line2D

legend_handles = [
    Line2D(
        [0], [0],
        color=color,
        marker="o",
        linewidth=2,
        markersize=6,
        label=category
    )
    for category, color in category_colors.items()
    if category in forest_df["Category"].unique()
]

fig.legend(
    handles=legend_handles,
    loc="lower center",
    ncol=4,
    frameon=False,
    bbox_to_anchor=(0.5, -0.015),
    fontsize=9
)

# ============================================================
# 11. FINAL LAYOUT AND EXPORT
# ============================================================

plt.subplots_adjust(
    left=0.22,
    right=0.98,
    top=0.92,
    bottom=0.12,
    wspace=0.08
)

plt.savefig(
    "Congo_Basin_Trophic_Group_Forest_Plot.png",
    dpi=600,
    bbox_inches="tight",
    facecolor="white"
)

plt.savefig(
    "Congo_Basin_Trophic_Group_Forest_Plot.pdf",
    bbox_inches="tight",
    facecolor="white"
)

plt.show()
```

```python

import numpy as np
import pandas as pd
import geopandas as gpd
import matplotlib.pyplot as plt
import matplotlib as mpl
import seaborn as sns

from matplotlib.colors import Normalize, TwoSlopeNorm

# ============================================================
# 1. MODEL CONFIGURATION
# ============================================================

grid = gpd.read_file('data/grid_overall_biodiversity.gpkg')
models = {
    "Carnivores": {
        "model": carnivore_richness_models['mosaic_vegetation_richness'],
        "response": "richness"
    },
    "Herbivores": {
        "model": herbivore_richness_models['mosaic_vegetation_richness'],
        "response": "richness"
    },
    "Omnivores": {
        "model": omnivore_richness_models['mosaic_vegetation_richness'],
        "response": "richness"
    }
}

taxon_order = [
    "Carnivores",
    "Herbivores",
    "Omnivores"
]

# ============================================================
# 2. GENERAL FIGURE STYLE
# ============================================================

sns.set_theme(style="white")

plt.rcParams.update({
    "font.family": "DejaVu Sans",
    "font.size": 10,
    "axes.titlesize": 12,
    "axes.titleweight": "bold",
    "figure.titlesize": 19,
    "savefig.dpi": 600
})

# ============================================================
# 3. EXTRACT OBSERVED, PREDICTED AND PEARSON RESIDUALS
# ============================================================

map_data = {}

for taxon in taxon_order:

    model = models[taxon]["model"]
    response = models[taxon]["response"]

    # Original data retained by the formula model
    model_data = model.model.data.frame

    # Original observations used by the fitted model
    row_index = model.model.data.row_labels

    if row_index is None:
        raise ValueError(
            f"{taxon}: model row indices are unavailable. "
            "Check the original model data alignment."
        )

    if not grid.index.is_unique:
        raise ValueError(
            "grid_gdf must have a unique index."
        )

    missing_rows = row_index.difference(grid.index)

    if len(missing_rows) > 0:
        raise ValueError(
            f"{taxon}: {len(missing_rows)} model rows "
            "are missing from grid_gdf."
        )

    # Observed response
    observed = model_data.loc[row_index, response].astype(float)

    # Model predictions
    predictions = np.asarray(
        model.predict(),
        dtype=float
    ).reshape(-1)

    if len(predictions) != len(row_index):
        raise ValueError(
            f"{taxon}: prediction length does not match model rows."
        )

    predicted = pd.Series(
        predictions,
        index=row_index
    )

    # Negative Binomial NB2 dispersion parameter
    if "alpha" not in model.params.index:
        raise ValueError(
            f"{taxon}: alpha parameter not found. "
            "Check that the model is a Negative Binomial NB2 model."
        )

    alpha = float(model.params["alpha"])

    if not np.isfinite(alpha) or alpha < 0:
        raise ValueError(
            f"{taxon}: invalid Negative Binomial alpha."
        )

    # Variance: Var(Y) = mu + alpha * mu^2
    variance = predicted + alpha * predicted**2

    # Pearson residuals
    pearson_residual = (
        observed - predicted
    ) / np.sqrt(variance)

    # Build spatial dataset
    spatial = grid.loc[
        row_index, ["geometry"]
    ].copy()

    spatial["Observed"] = observed.values
    spatial["Predicted"] = predicted.values
    spatial["Raw_residual"] = (
        observed.values - predicted.values
    )
    spatial["Pearson_residual"] = pearson_residual.values

    map_data[taxon] = spatial

# ============================================================
# 4. DEFINE SHARED COLOUR SCALES
# ============================================================

# Shared richness scale across all six observed/predicted maps
all_richness = np.concatenate([
    np.concatenate([
        map_data[taxon]["Observed"].to_numpy(),
        map_data[taxon]["Predicted"].to_numpy()
    ])
    for taxon in taxon_order
])

# 99th percentile reduces the influence of extreme values
richness_vmax = np.nanpercentile(all_richness, 99)

if not np.isfinite(richness_vmax) or richness_vmax <= 0:
    richness_vmax = np.nanmax(all_richness)

richness_norm = Normalize(
    vmin=0,
    vmax=richness_vmax
)

# Shared, symmetric residual scale
all_residuals = np.concatenate([
    map_data[taxon]["Pearson_residual"].to_numpy()
    for taxon in taxon_order
])

residual_limit = np.nanpercentile(
    np.abs(all_residuals),
    98
)

if not np.isfinite(residual_limit) or residual_limit <= 0:
    residual_limit = np.nanmax(np.abs(all_residuals))

if not np.isfinite(residual_limit) or residual_limit <= 0:
    residual_limit = 1

residual_norm = TwoSlopeNorm(
    vmin=-residual_limit,
    vcenter=0,
    vmax=residual_limit
)

# ============================================================
# 5. CREATE FIGURE
# ============================================================

fig, axes = plt.subplots(
    nrows=3,
    ncols=3,
    figsize=(19, 15)
)

# Map colour palettes
richness_cmap = "YlGnBu"
residual_cmap = "RdBu_r"

column_titles = [
    "A. Observed species richness",
    "B. Predicted species richness",
    "C. Pearson residuals"
]

# Column headings
for j, title in enumerate(column_titles):

    axes[0, j].set_title(
        title,
        fontsize=13,
        fontweight="bold",
        pad=15
    )

# ============================================================
# 6. DRAW SPATIAL MAPS
# ============================================================

for i, taxon in enumerate(taxon_order):

    spatial = map_data[taxon]

    # --------------------------------
    # Column 1: Observed richness
    # --------------------------------

    spatial.plot(
        column="Observed",
        ax=axes[i, 0],
        cmap=richness_cmap,
        norm=richness_norm,
        linewidth=0,
        edgecolor="none",
        legend=False,
        missing_kwds={
            "color": "lightgrey",
            "label": "Missing data"
        }
    )

    # --------------------------------
    # Column 2: Predicted richness
    # --------------------------------

    spatial.plot(
        column="Predicted",
        ax=axes[i, 1],
        cmap=richness_cmap,
        norm=richness_norm,
        linewidth=0,
        edgecolor="none",
        legend=False,
        missing_kwds={
            "color": "lightgrey",
            "label": "Missing data"
        }
    )

    # --------------------------------
    # Column 3: Pearson residuals
    # --------------------------------

    spatial.plot(
        column="Pearson_residual",
        ax=axes[i, 2],
        cmap=residual_cmap,
        norm=residual_norm,
        linewidth=0,
        edgecolor="none",
        legend=False,
        missing_kwds={
            "color": "lightgrey",
            "label": "Missing data"
        }
    )

    # Taxonomic group label
    axes[i, 0].set_ylabel(
        taxon,
        fontsize=13,
        fontweight="bold",
        labelpad=12
    )

    # Remove axes and geographic ticks
    for j in range(3):

        axes[i, j].set_axis_off()
        axes[i, j].set_aspect("equal")

# ============================================================
# 7. SHARED COLOUR BAR: OBSERVED AND PREDICTED
# ============================================================

richness_sm = mpl.cm.ScalarMappable(
    norm=richness_norm,
    cmap=richness_cmap
)

richness_sm.set_array([])

cbar_ax1 = fig.add_axes([
    0.18, 0.075, 0.43, 0.018
])

cbar1 = fig.colorbar(
    richness_sm,
    cax=cbar_ax1,
    orientation="horizontal"
)

cbar1.set_label(
    "Species richness (number of species)",
    fontsize=11,
    fontweight="bold",
    labelpad=7
)

# ============================================================
# 8. SHARED COLOUR BAR: RESIDUALS
# ============================================================

residual_sm = mpl.cm.ScalarMappable(
    norm=residual_norm,
    cmap=residual_cmap
)

residual_sm.set_array([])

cbar_ax2 = fig.add_axes([
    0.70, 0.075, 0.22, 0.018
])

cbar2 = fig.colorbar(
    residual_sm,
    cax=cbar_ax2,
    orientation="horizontal"
)

cbar2.set_label(
    "Pearson residual",
    fontsize=11,
    fontweight="bold",
    labelpad=7
)

# ============================================================
# 9. MAIN TITLE AND SUBTITLE
# ============================================================

fig.suptitle(
    "Observed and predicted species richness "
    "across trophic groups in the Congo Basin",
    fontsize=19,
    fontweight="bold",
    y=0.975
)

fig.text(
    0.5,
    0.945,
    "Negative Binomial regression models | "
    "Observed richness, model predictions and Pearson residuals",
    ha="center",
    fontsize=11,
    color="dimgray"
)

# ============================================================
# 10. FOOTNOTE
# ============================================================

fig.text(
    0.5,
    0.025,
    "Pearson residual = (observed − predicted) / "
    "√(predicted + α × predicted²). "
    "Positive values indicate underprediction; "
    "negative values indicate overprediction.",
    ha="center",
    fontsize=9,
    color="dimgray"
)

# ============================================================
# 11. LAYOUT AND EXPORT
# ============================================================

plt.subplots_adjust(
    left=0.08,
    right=0.96,
    top=0.90,
    bottom=0.13,
    hspace=0.08,
    wspace=0.04
)

plt.savefig(
    "Congo_Basin_Trophic_Richness_Spatial_Maps.png",
    dpi=600,
    bbox_inches="tight",
    facecolor="white"
)

plt.savefig(
    "Congo_Basin_Trophic_Richness_Spatial_Maps.pdf",
    bbox_inches="tight",
    facecolor="white"
)

plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# --------------------------------------------------
# 1. Define your three fitted models
# --------------------------------------------------

models = {
    "Carnivores": carnivore_richness_models['mosaic_vegetation_richness'],
    "Herbivores": herbivore_richness_models['mosaic_vegetation_richness'],
    "Omnivores": omnivore_richness_models['mosaic_vegetation_richness']
}

# --------------------------------------------------
# 2. Define predictor groups
# Use the exact term names from your model formulas.
# C(ORD_STRA) represents a categorical predictor.
# --------------------------------------------------

predictor_groups = {
    "Climate": [
        "BIO1_temporal_mean",
        "BIO1_temporal_SD",
        "BIO12_temporal_mean",
        "BIO12_temporal_SD"
    ],

    "Vegetation": [
        "vegetation_richness",
        "vegetation_shannon",
        "vegetation_evenness"
    ],

    "Topography": [
        "slope_std",
        "TPI_mean",
        "roughness_mean",
        "roughness_std",
        "hillshade_mean",
        "hillshade_std"
    ],

    "Hydrology": [
        "C(ORD_STRA)",
        "distance_km",
        "LENGTH_KM",
        "DIS_AV_CMS"
    ],

    "Forest structure": [
        "forest_canopy_mean",
        "forest_canopy_stdev",
        "tree_cover_mean",
        "tree_cover_stdev",
        "forest_loss_mean"
    ],

    "Soil": [
        "clay_mean",
        "clay_stdev",
        "cec_mean",
        "cec_stdev",
        "soc_mean",
        "soc_stdev",
        "ph_mean",
        "ph_stdev"
    ],

    "Palaeogeography": [
        "refugia_dist_km"
    ]
}

# --------------------------------------------------
# 3. Calculate joint Wald statistics
# --------------------------------------------------

def calculate_joint_wald(model, predictor_groups):

    params = model.params
    covariance = model.cov_params()

    # Exclude the dispersion parameter
    valid_params = [
        p for p in params.index
        if p.lower() != "alpha"
        and p.lower() != "const"
        and p.lower() != "intercept"
    ]

    results = []

    for group, variables in predictor_groups.items():

        selected_params = []

        for variable in variables:

            for param in valid_params:

                # Match ordinary and categorical coefficients
                if (
                    param == variable
                    or param.startswith(variable + "[")
                    or param.startswith(variable + ":")
                    or param.startswith("C(" + variable + ")[")
                    or param.startswith("C(" + variable + ",")
                ):
                    selected_params.append(param)

        selected_params = list(dict.fromkeys(selected_params))

        if len(selected_params) == 0:
            continue

        beta = params.loc[selected_params].to_numpy()

        cov = covariance.loc[
            selected_params,
            selected_params
        ].to_numpy()

        # Joint Wald statistic
        wald = float(
            beta.T @ np.linalg.pinv(cov) @ beta
        )

        results.append({
            "Predictor": group,
            "Wald": wald,
            "DF": len(selected_params)
        })

    result = pd.DataFrame(results)

    if result.empty or result["Wald"].sum() == 0:
        raise ValueError("No valid predictor groups were identified.")

    result["Relative importance (%)"] = (
        result["Wald"] / result["Wald"].sum()
    ) * 100

    return result


# --------------------------------------------------
# 4. Calculate importance for all three models
# --------------------------------------------------

all_results = []

for taxon, model in models.items():

    if isinstance(model, dict):
        model = model["model"]

    importance = calculate_joint_wald(
        model,
        predictor_groups
    )

    importance["Trophic group"] = taxon

    all_results.append(importance)

importance_df = pd.concat(
    all_results,
    ignore_index=True
)

importance_df["JWI_per_DF"] = (
    importance_df["Relative importance (%)"]
    / importance_df["DF"]
)

print(
    importance_df.sort_values(
        ["Trophic group", "Relative importance (%)"],
        ascending=[True, False]
    )
)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Pivot data into heatmap format
heatmap_data = importance_df.pivot(
    index="Predictor",
    columns="Trophic group",
    values="JWI_per_DF"
)

# Set column order
heatmap_data = heatmap_data[
    ["Carnivores", "Herbivores", "Omnivores"]
]

# Sort predictor groups by average importance
heatmap_data = heatmap_data.loc[
    heatmap_data.mean(axis=1).sort_values().index
]

print(heatmap_data.round(2))

sns.set_theme(style="white", context="paper")

fig, ax = plt.subplots(figsize=(10, 8))

sns.heatmap(
    heatmap_data,
    annot=True,
    fmt=".1f",
    cmap="YlOrRd",
    linewidths=1,
    linecolor="white",
    vmin=0,
    vmax=heatmap_data.max().max(),
    cbar_kws={
        "label": "Relative JWI per degree of freedom ()",
        "shrink": 0.8
    },
    annot_kws={
        "size": 11,
        "weight": "bold"
    },
    ax=ax
)

ax.set_title(
    "Average Relative Wald Importance per Coefficient",
    fontsize=15,
    fontweight="bold",
    pad=18
)

ax.set_xlabel("Trophic group", fontsize=12, labelpad=10)
ax.set_ylabel("Predictor group", fontsize=12, labelpad=10)

ax.tick_params(axis="both", labelsize=11)

plt.tight_layout()

plt.savefig(
    "JWI_heatmap_trophic_groups.png",
    dpi=400,
    bbox_inches="tight"
)

plt.show()
```
