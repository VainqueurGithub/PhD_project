```python
import pandas as pd
import numpy as np
import geopandas as gpd
import rasterio
import shapely
import pyproj
import xarray as xr
import fiona
import pyogrio
import matplotlib as plt
import geodatasets
import folium
from shapely.geometry import Polygon
from shapely.geometry import box
import matplotlib.cm as cm
import matplotlib.colors as colors
```

```python
fiona.listlayers("data/Congobasin_perimeter/SPCB_Report_Perimeter.gdb")
```

```python
congo_basin_boundries = gpd.read_file("data/Congobasin_perimeter/SPCB_Report_Perimeter.gdb", layer="SPCB_Boundary")
congo_basin_boundries.to_crs(epsg='4326')
congo_mammals = gpd.read_file('congo_mammals.gpkg').to_crs(epsg='4326')
```

```python
congo_mammals[congo_mammals["sci_name"]=='Gorilla gorilla']
```

```python
surrogates = congo_mammals[
    congo_mammals["sci_name"].isin([
        "Cercopithecus ascanius","Cercopithecus cephus","Cercopithecus nictitans","Cercopithecus neglectus","Cercopithecus wolfi",
        "Cercopithecus pogonias","Cercopithecus mona","Pan troglodytes","Pan paniscus","Gorilla gorilla","Loxodonta cyclotis","Panthera pardus",
        "Caracal aurata","Crocuta crocuta","Syncerus caffer nanus","Potamochoerus porcus","Tragelaphus spekii gratus","Philantomba monticola",
        "Cephalophus callipygus","Cephalophus nigrifrons","Cephalophus dorsalis","Cephalophus leucogaster","Cephalophus silvicultor",
        "Cephalophus ogilbyi","Tragelaphus scriptus","Tragelaphus sylvaticus"
    ])
]

surrogates
```

```python
# Mammals biodiversity priority map. This map show where mammal surrogate species overlap

#Create a grid over Congo Basin
xmin, ymin, xmax, ymax = congo_mammals.total_bounds
cell_size = 0.25  # degrees (adjust later: 0.5, 0.25, etc.)

grid_cells = [
    box(x, y, x + cell_size, y + cell_size)
    for x in np.arange(xmin, xmax, cell_size)
    for y in np.arange(ymin, ymax, cell_size)
]

grid = gpd.GeoDataFrame(geometry=grid_cells, crs="EPSG:4326")

#Count surrogate occurrence per cell
grid["species"] = None

for i, cell in grid.iterrows():
    inside = surrogates[surrogates.intersects(cell.geometry)]
    grid.at[i, "species"] = list(inside["sci_name"].unique())
    
# Computer Richness
grid["richness"] = grid["species"].apply(len)
grid["species"] = grid["species"].apply(lambda x: ", ".join(x) if x else "")
#Color map in Folium
m = folium.Map(location=[0, 20], zoom_start=4, tiles="CartoDB positron")

colormap = cm.get_cmap("viridis")
max_richness = grid["richness"].max()

def get_color(r):

    if r == 0:
        return "transparent"

    norm = r / max_richness
    rgba = colormap(norm)

    # convert to hex
    return colors.rgb2hex(rgba)

def style_function(feature):

    r = feature["properties"]["richness"]

    if r == 0:
        return {
            "fillOpacity": 0,
            "weight": 0
        }

    return {
        "fillColor": get_color(r),
        "color": "black",
        "weight": 0.1,
        "fillOpacity": 0.7
    }

#Add to map
folium.GeoJson(
    grid,
    style_function=style_function,
    tooltip=folium.GeoJsonTooltip(fields=["species","richness"], aliases=["Species present","Surrogate richness"])
).add_to(m)

m.save("mammal_surrogate_richness_map.html")
```

```python
# Mammals biodiversity priority map. This map show where mammal surrogate species overlap with other mammals

#Create a grid over Congo Basin
xmin, ymin, xmax, ymax = congo_mammals.total_bounds
cell_size = 0.25  # degrees (adjust later: 0.5, 0.25, etc.)

grid_cells = [
    box(x, y, x + cell_size, y + cell_size)
    for x in np.arange(xmin, xmax, cell_size)
    for y in np.arange(ymin, ymax, cell_size)
]

grid = gpd.GeoDataFrame(geometry=grid_cells, crs="EPSG:4326")

#Count surrogate occurrence per cell
grid["species"] = None
grid["surrogates"] = None

for i, cell in grid.iterrows():
    inside = congo_mammals[congo_mammals.intersects(cell.geometry)]
    grid.at[i, "species"] = list(inside["sci_name"].unique())

for i, cell in grid.iterrows():
    inside = surrogates[surrogates.intersects(cell.geometry)]
    grid.at[i, "surrogates"] = list(inside["sci_name"].unique())
# Computer Richness
grid["richness"] = grid["species"].apply(len)
grid["surrogates"] = grid["surrogates"].apply(lambda x: ", ".join(x) if x else "")
#Color map in Folium
m = folium.Map(location=[0, 20], zoom_start=4, tiles="CartoDB positron")

colormap = cm.get_cmap("viridis")
max_richness = grid["richness"].max()

def get_color(r):

    if r == 0:
        return "transparent"

    norm = r / max_richness
    rgba = colormap(norm)

    # convert to hex
    return colors.rgb2hex(rgba)

def style_function(feature):

    r = feature["properties"]["richness"]

    if r == 0:
        return {
            "fillOpacity": 0,
            "weight": 0
        }

    return {
        "fillColor": get_color(r),
        "color": "black",
        "weight": 0.1,
        "fillOpacity": 0.7
    }

#Add to map
folium.GeoJson(
    grid,
    style_function=style_function,
    tooltip=folium.GeoJsonTooltip(fields=["surrogates","richness"], aliases=["Surrogates present","Species richness"])
).add_to(m)

m.save("mammals_richness_map.html")
```
